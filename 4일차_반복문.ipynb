## 🔁 for문 - 정해진 횟수만큼

for문은 **정해진 범위나 리스트의 요소들을 순회하며 하나씩 처리**할 때 사용합니다.

````python
# 환자 명단 순회
patient_names = ["김환자", "이환자", "박환자", "최환자"]  # 오늘 진료 환자 명단 요소(콤마를 기준으로 나뉘는 데이터)의 수가 4개

# for문에서 정의된 변수명은 for문 안에서만 사용되고, 밖에서는 사용되지 못한다. 이 때 변수명의 타입은 할당되는 요소의 타입이다.
for patient_name in patient_names:
    print(f"{patient_name}님, 진료실로 들어오세요!")

print("\n======= 두 번 반복 =======")
# 만약 리스트를 두 번 순회하고 싶다면?
for patient_name in patient_names * 2:
    print(f"{patient_name}님, 진료실로 들어오세요!")
````

반복문이 없다면

````python
# ---- loop 1 ----#
print(patient_names[0])

# ---- loop 1 ----#
print(patient_names[1])

# ---- loop 1 ----#
print(patient_names[2])

# ---- loop 1 ----#
print(patient_names[3])
````

## 🔍 유용한 반복문 함수들

| 함수 | 용도 | 반환값 | 예시 |
| --- | --- | --- | --- |
| `range()` | 연속된 숫자 생성 | 숫자 시퀀스 | `range(5)` → 0,1,2,3,4 |
| `enumerate()` | 인덱스와 값 함께 | (인덱스, 값) 튜플 | `enumerate(['a','b'])` → (0,'a'), (1,'b') |
| `zip()` | 여러 리스트 동시 순회 | 각 리스트의 요소들을 묶은 튜플 | `zip([1,2], ['a','b'])` → (1,'a'), (2,'b') |



### enumerate() - 인덱스와 값을 함께

`enumerate()`는 **인덱스(순서)와 값을 함께** 가져올 때 사용합니다.



````python
# 기본 사용법 - 환자 대기 순번
patients = ["김환자", "이환자", "박환자", "최환자"]  # 대기 환자 명단

for index, patient_name in enumerate(patients):
    print(f"{index + 1}번째 대기: {patient_name}")
````

````python
# 시작 번호 지정하기 - 병실 번호 배정
patients = ["김환자", "이환자", "박환자"]  # 입원 환자 명단

print("=== 병실 배정 현황 ===")
for room_number, patient_name in enumerate(patients, 101):  # 101호부터 시작
    print(f"{room_number}호: {patient_name}")
````

````python
# 활용하기 - 처방전 목록
prescriptions = ["해열제", "항생제", "진통제", "비타민"]  # 처방 약물 목록

print("오늘의 처방전")
for num, medicine in enumerate(prescriptions, start=1):
    print(f"{num}. {medicine}")
````

### zip() - 여러 리스트를 동시에

`zip()`은 **여러 개의 리스트를 동시에 순회**할 때 사용합니다.




````python
# 기본 사용법 - 환자 기본 정보
patient_names = ["김환자", "이환자", "박환자"]  # 환자 이름 목록
patient_ages = [65, 45, 32]  # 환자 나이 목록
departments = ["내과", "외과", "소아과"]  # 담당 진료과 목록

for data in zip(patient_names, patient_ages, departments):
    print(data) # 튜플로 묶임
````

````python
# 튜플로 묶여있는 데이터를 언패킹으로 각각 꺼내오기
for name, age, department in zip(patient_names, patient_ages, departments):
    print(f"{name}님({age}세)은 {department}에서 진료받습니다.")
````

````python
# 활용하기 - 검사 항목과 결과
test_items = ["혈압", "혈당", "콜레스테롤", "체중"]  # 검사 항목명
test_results = [120, 95, 180, 70]  # 검사 결과값
units = ["mmHg", "mg/dL", "mg/dL", "kg"]  # 측정 단위

print("🔬 검사 결과")
for item, result, unit in zip(test_items, test_results, units):
    print(f"{item}: {result} {unit}")
````

### range() 함수 활용

range() 함수는 **연속된 숫자를 생성**하는 함수입니다.

**range() 함수의 인자**

- `range(끝)`: 0부터 끝-1까지
- `range(시작, 끝)`: 시작부터 끝-1까지
- `range(시작, 끝, 간격)`: 시작부터 끝-1까지 간격만큼 증가



#### 리스트로 변환하기




````python
# range(끝): 0부터 끝-1까지 병실 번호 생성
room_numbers = list(range(5))  # 0~4번 격리실
print(f"격리실 번호: {room_numbers}")

# tip: ctrl + D or command + D를 사용하면 한번에 여러 변수명 수정 가능
````

````python
# range(시작, 끝): 시작부터 끝-1까지 일반 병실 번호
general_rooms = list(range(101, 111))  # 101~110호 일반실
print(f"일반실 번호: {general_rooms}")
````

````python
# range(시작, 끝, 간격): 간격만큼 증가하는 VIP실 번호
vip_rooms = list(range(201, 211, 2))  # 201, 203, 205, 207, 209호
print(f"VIP실 번호: {vip_rooms}")
````

````python
# 거꾸로 카운트 - 수술 대기 순서
surgery_queue = list(range(10, 0, -1))  # 10번부터 1번까지
print(f"수술 대기 순서: {surgery_queue}")
````

## ♾️ while문 - 조건이 맞을 때까지

while문은 **조건이 참인 동안 계속 반복**합니다.


````python
# 기본 while문
count = 1

while count <= 5:
    print(f"현재 숫자: {count}")
    count += 1  # count = count + 1 과 같음

print("카운트 완료!")
````

````python
num = 0

# num = num + 1
num += 1

print(num)
````

활용하기

````python
# 혈압 안정화 대기
systolic_pressure = 160  # 현재 수축기 혈압
target_pressure = 140  # 목표 혈압
hour_count = 0  # 경과 시간

print("=== 혈압 안정화 치료 ===")
while systolic_pressure > target_pressure:
    hour_count += 1
    systolic_pressure -= 5  # 시간당 5씩 감소 가정
    print(f"{hour_count}시간 후: {systolic_pressure} mmHg")

print(f"혈압 안정화 완료! (총 {hour_count}시간 소요)")
````

## 🛑 break와 continue

- **break**: 반복문을 완전히 종료
- **continue**: 현재 반복을 건너뛰고 다음 반복으로


````python
# break - 특정 조건에서 반복 종료
for i in range(1, 11):
    if i == 5:
        print("5를 만났으니 종료!")
        break  # 반복문 완전 종료
    print(f"숫자: {i}")
````

````python
for i in range(1, 11):
    print(f"숫자: {i}")
    break  # 반복문 완전 종료
````

````python
# continue - 특정 조건에서 건너뛰기
for i in range(1, 6):
    if i == 3:
        print("3은 건너뛰기!")
        continue  # 3일 때만 아래 코드 실행 안함
    print(f"숫자: {i}")
````

활용하기

````python
# 통합 환자 데이터
patient_data = {
    "김환자": {"systolic": 120, "allergic": False},  # 수축기 혈압, 알레르기 여부
    "이환자": {"systolic": 145, "allergic": True},
    "박환자": {"systolic": 160, "allergic": False},
    "최환자": {"systolic": 135, "allergic": True},
    "정환자": {"systolic": 180, "allergic": False}
}
````

````python
# break - 위험 혈압 환자 첫 발견시 즉시 응급 처치
danger_threshold = 170  # 위험 혈압 기준값

print("=== 응급 혈압 환자 검색 ===")
for patient_name, data in patient_data.items():
    systolic = data["systolic"]  # 수축기 혈압
    print(f"{patient_name} 혈압 확인: {systolic} mmHg")

    if systolic >= danger_threshold:
        print(f"🚨 위험! {patient_name}의 혈압이 {systolic} mmHg입니다!")
        print("즉시 응급 처치를 시작합니다.")
        break  # 위험 환자 발견시 즉시 중단

    print(f"{patient_name} 정상 범위")

print("\n응급 혈압 환자 검색을 종료합니다.")
````

````python
print("=== 페니실린 처방 가능 환자 ===")
# continue 예제 - 알레르기 환자 제외하고 처방
for patient_name, data in patient_data.items():
    if data["allergic"]:  # 알레르기 여부 확인
        print(f"{patient_name}: 알레르기로 처방 제외")
        continue  # 알레르기 환자는 건너뛰기

    systolic = data["systolic"]
    print(f"{patient_name}: 페니실린 처방 가능 (혈압: {systolic} mmHg)")
````

## ❓문제 6. 응급환자 발견

환자들의 체온을 순서대로 확인하다가 38.5도 이상인 고열 환자를 발견하면
"응급환자 발견! 즉시 처치 필요" 출력하고 반복을 중단하세요.

## 💊 실전 예제: 환자 검사 결과 분석 시스템



````python
# 환자 검사 결과 데이터
patient_results = [
    {"name": "김환자", "age": 65, "systolic": 145, "blood_sugar": 110},  # 환자별 검사 결과
    {"name": "이환자", "age": 45, "systolic": 130, "blood_sugar": 95},
    {"name": "박환자", "age": 32, "systolic": 120, "blood_sugar": 85},
    {"name": "최환자", "age": 70, "systolic": 160, "blood_sugar": 200}
]
````

````python
# 1. 모든 환자 기본 정보 출력
for patient in patient_results:
    name = patient["name"]  # 환자 이름
    age = patient["age"]  # 환자 나이
    systolic = patient["systolic"]  # 수축기 혈압
    blood_sugar = patient["blood_sugar"]  # 혈당 수치
    print(f"{name}({age}세): 혈압 {systolic}mmHg, 혈당 {blood_sugar}mg/dL")
````

````python
# 2. 고위험군 환자 찾기
high_risk_patients = []  # 고위험군 환자 이름 목록

for patient in patient_results:
    name = patient["name"]
    age = patient["age"]
    systolic = patient["systolic"]
    blood_sugar = patient["blood_sugar"]

    # 위험 요소 확인
    is_hypertension = systolic >= 140  # 고혈압 여부
    is_diabetes = blood_sugar >= 140  # 당뇨 여부
    is_elderly = age >= 65  # 고령자 여부

    # 2개 이상 조건 만족시 고위험군
    risk_factors = [is_hypertension, is_diabetes, is_elderly]
    if sum(risk_factors) >= 2: # sum: True의 갯수 합산
        high_risk_patients.append(name)

# 고위험군 환자 순번 출력
if high_risk_patients:
    for rank, patient_name in enumerate(high_risk_patients, 1):
        print(f"{rank}순위 고위험군: {patient_name}")
else:
    print("고위험군 환자가 없습니다.")
````

````python
# 3. 검사 항목별 평균 계산 (리스트 컴프리헨션 활용)
systolic_values = [patient["systolic"] for patient in patient_results]  # 모든 혈압 수치
blood_sugar_values = [patient["blood_sugar"] for patient in patient_results]  # 모든 혈당 수치

avg_systolic = sum(systolic_values) / len(systolic_values)  # 평균 혈압
avg_blood_sugar = sum(blood_sugar_values) / len(blood_sugar_values)  # 평균 혈당

print(f"평균 수축기 혈압: {avg_systolic:.1f} mmHg")
print(f"평균 혈당: {avg_blood_sugar:.1f} mg/dL")
````

````python
# 4. 정상 범위 환자 비율
normal_bp_patients = [patient["name"] for patient in patient_results if patient["systolic"] < 140]
normal_bs_patients = [patient["name"] for patient in patient_results if patient["blood_sugar"] < 140]

print(f"\n정상 혈압 환자: {len(normal_bp_patients)}명 ({normal_bp_patients})")
print(f"정상 혈당 환자: {len(normal_bs_patients)}명 ({normal_bs_patients})")
````

# 퀴즈: 비만도(BMI) 분석 및 상담 시스템

건강검진 센터에서 회원들의 키와 몸무게 데이터를 기반으로 비만도(BMI)를 계산하고, 건강 상담 리포트를 출력하는 프로그램을 작성해보세요.

## 📝 요구사항

1.  **BMI 계산 공식**
    * $BMI = 몸무게(kg) / (키(m) x 키(m))$
    * *주의: 데이터의 키는 cm 단위이므로 m 단위로 변환해야 합니다.*

2.  **비만도 분류 기준 (대한비만학회 기준)**
    * **저체중**: BMI < 18.5
    * **정상**: 18.5 ≤ BMI < 23
    * **과체중**: 23 ≤ BMI < 25
    * **비만**: BMI ≥ 25

4.  **출력 형식 예시**
```text
=== 회원별 건강 리포트 ===
김철수(175cm, 75kg): BMI 24.5 - 과체중 (운동 필요)
이영희(160cm, 45kg): BMI 17.6 - 저체중 (영양 섭취 요망)
박민수(180cm, 95kg): BMI 29.3 - 비만 (정밀 검사 권장)


````python
# BMI 계산기 만들기

# 기본 데이터 - 회원 신체 정보
members = [
    {"name": "김철수", "height": 175, "weight": 75},
    {"name": "이영희", "height": 160, "weight": 45},
    {"name": "박민수", "height": 180, "weight": 95},
    {"name": "최정상", "height": 172, "weight": 65},
    {"name": "정건강", "height": 168, "weight": 60}
]

print("=== 회원별 건강 리포트 ===")

for member in members:
  name, height, weight = member.values()
  height_m = height / 100
  bmi = weight / (height_m ** 2)

  result = {}
  if bmi < 18.5:
    result["step"] = "저체중"
    result["text"] = "영양 섭취 요망"
  elif bmi < 23:
    result["step"] = "정상"
    result["text"] = "건강합니다!"
  elif bmi < 25:
    result["step"] = "과체중"
    result["text"] = "운동 필요"
  else:
    result["step"] = "비만"
    result["text"] = "정밀 검사 권장"

  if result["step"] != "정상":
    print(f"{name}({height}cm, {weight}kg): BMI {bmi:.1f} - {result["step"]} ({result["text"]})")
````
