# 3. 조건문

## 🤔 if문 - 상황에 따라 다르게

조건문은 **주어진 조건에 따라 다른 코드를 실행**하는 구문입니다.  
의료 현장에서는 환자의 상태나 검사 결과에 따라 다른 처치를 결정해야 하므로 조건문이 매우 중요합니다.

조건문은 조건식과 실행구문으로 이루어져 있습니다.

```python
# 조건문 기분 구문
if 조건식:
  # 조건식이 True인 경우 실행할 구문
  print("조건식이 True입니다")
```

## 🔵 bool 자료형과 조건문

bool(불린) 자료형은 참(True) 또는 거짓(False) 두 가지 값만 가집니다.  
조건문의 핵심이 되는 자료형입니다. **`조건식`에는 무조건 `bool`자료형이 들어가야 하기 때문입니다.**


````python
# ❌ 잘못된 예시 - 대입 연산자 사용
patient_age = 25

if patient_age = 18:  # SyntaxError!
    print("성인이 되었습니다!")
````

````python
# ✅ 올바른 예시 - 비교 연산자 사용
patient_age = 25

if patient_age == 18:
    print("성인이 되었습니다!")
````

### 조건문에서 bool 값 활용
bool 변수는 조건문에서 직접 사용할 수 있습니다. `== True`를 쓸 필요가 없습니다.


````python
# ❌ 잘못된 예시 - bool 변수에 불필요한 비교
is_emergency = True
if is_emergency == True:
  print("응급상황입니다!")
````

````python
# ✅ 올바른 예시 - bool 변수 직접 사용
if is_emergency:
    print("응급상황입니다!")
````

````python
# 환자 상태에 따른 처리
is_emergency = True
has_insurance = False

if is_emergency:  # True이면 실행
    print("응급실로 이동하세요!")

if not has_insurance:  # False이면 실행 (not 사용)
    print("보험 정보를 확인해주세요")
````

### 비교 연산의 결과는 bool
모든 비교 연산의 결과는 bool 자료형입니다.


````python
# 검사 결과 평가
blood_sugar = 180
is_diabetes = blood_sugar >= 140  # 결과: True

cholesterol = 220
is_high_cholesterol = cholesterol >= 240  # 결과: False

print(f"당뇨 진단: {is_diabetes}")
print(f"고콜레스테롤혈증: {is_high_cholesterol}")

````

## 🎯 조건문의 기본 구조

### If A: 만약에 A가 True라면
가장 기본적인 조건문 형태입니다. 조건이 참(True)일 때만 코드가 실행됩니다.

```python
if bool:
    # 실행문
    print("조건식이 True인경우 실행할 로직")
```

````python
# 환자 체온 확인
temperature = 38.5

if temperature >= 37.5:
    print("발열이 있습니다. 해열제를 처방하겠습니다.")
````

### if-else - 만약에 A가 False라면
조건이 참일 때와 거짓일 때 각각 다른 코드를 실행합니다.

````python
# 환자 나이에 따른 처방
patient_age = 16

if patient_age >= 18:
    print("성인용 약물을 처방합니다.")
else:
    print("소아용 약물을 처방합니다.")
````

### if-elif-else - 여러 조건 처리
여러 개의 조건을 순서대로 확인하여 해당하는 조건의 코드만 실행합니다.

````python
# BMI에 따른 체중 분류
bmi = 27.5

if bmi < 18.5:
    weight_status = "저체중"
elif bmi < 25:
    weight_status = "정상체중"
elif bmi < 30:
    weight_status = "과체중"
else:
    weight_status = "비만"

print(f"BMI {bmi}: {weight_status}")
````

## 🔍 조건식 연산자


### 비교 연산자
조건문에서 값들을 비교할 때 사용하는 연산자들입니다.


````python
# 환자 혈압 데이터
systolic = 140 # 환자의 수축기 혈압
normal_systolic = 120 # 정상 수축기 혈압 기준값

print(f"수축기 혈압 {systolic} > 정상 {normal_systolic}: {systolic > normal_systolic}")
print(f"혈압 정상여부: {systolic <= normal_systolic}")
print(f"고혈압 진단: {systolic >= 140}")
print(f"정상 혈압: {systolic == normal_systolic}")
print(f"혈압 이상: {systolic != normal_systolic}")
````

### 포함 연산자 (in, not in)
리스트나 문자열에 특정 값이 포함되어 있는지 확인할 때 사용합니다.

- `in`: 포함되어 있으면 True
- `not in`: 포함되지 않으면 True


````python
# 환자 알레르기 확인
patient_allergies = ["페니실린", "아스피린", "설파제"] # 환자의 알레르기 약물 목록
prescribed_drug = "페니실린" # 처방하려는 약물명

if prescribed_drug in patient_allergies:
    print(f"⚠️ 주의: {prescribed_drug}에 알레르기가 있습니다!")
else:
    print(f"✅ {prescribed_drug} 처방 가능합니다.")
````

````python
# 증상 확인
patient_symptoms = ["발열", "기침", "두통"] # 환자가 호소하는 증상 목록

if "호흡곤란" not in patient_symptoms:
    print("호흡곤란 증상은 없습니다.")
````

### 논리 연산자

논리 연산자는 **여러 조건을 조합**해서 더 복잡한 조건을 만들 때 사용합니다.

| 연산자 | 의미 | 설명 |
| --- | --- | --- |
| `and` | 그리고 | 모든 조건이 True일 때만 True |
| `or` | 또는 | 하나라도 True이면 True |
| `not` | 아니다 | True를 False로, False를 True로 |


#### and 연산자 - 모든 조건이 참일 때


````python
# 수술 가능 여부 판단 (모든 조건을 만족해야 함)
patient_age = 45 # 환자 나이
systolic = 130  # 환자의 수축기 혈압

if patient_age >= 18 and systolic < 180:
    print("✅ 수술 가능합니다.")
else:
    print("❌ 수술 전 추가 검사가 필요합니다.")
````

#### or 연산자 - 하나라도 참이면


````python
# 응급실 방문 권고 (하나라도 해당되면 응급상황)
chest_pain = True  # 가슴 통증 여부
difficulty_breathing = False  # 호흡곤란 여부
severe_headache = False  # 심한 두통 여부

if chest_pain or difficulty_breathing or severe_headache:
    print("🚨 즉시 응급실로 오세요!")
else:
    print("일반 진료 예약을 잡으시기 바랍니다.")
````

#### not 연산자 - 반대 조건


````python
# 처방 금기 확인
is_pregnant = False  # 임신 여부
is_allergic = False  # 해당 약물 알레르기 여부

if not is_pregnant and not is_allergic:
    print("✅ 해당 약물 처방 가능합니다.")
````

````python
# 백신 접종 가능 여부
has_fever = True # 발열 여부

if not has_fever:
    print("백신 접종이 가능합니다.")
else:
    print("발열이 있어 백신 접종을 연기합니다.")

````

#### 복합 조건 활용


````python
# 복잡한 진료 조건 판단
patient_age = 55  # 환자 나이
is_diabetic = True  # 당뇨병 진단 여부
blood_sugar = 200  # 현재 혈당 수치 (mg/dL)
last_checkup_months = 8  # 마지막 검진 후 경과 개월 수

# 당뇨 환자의 정기검진 필요성
needs_urgent_care = (is_diabetic and blood_sugar > 180) or (patient_age > 50 and last_checkup_months >= 6)
````

````python
if needs_urgent_care:
    print("🏥 빠른 시일 내에 진료를 받으시기 바랍니다.")

    if is_diabetic and blood_sugar > 180:
        print("혈당 수치가 높습니다.")

    if patient_age > 50 and last_checkup_months >= 6:
        print("정기검진 시기입니다.")
else:
    print("현재 상태로는 정기 진료만으로 충분합니다.")
````

## 🔀 조건문 처리 방식 비교

| 방식 | 사용 시점 | 특징 | 예시 |
| --- | --- | --- | --- |
| **다중 조건** (if-elif-else) | 하나만 선택해야 할 때 | 첫 번째 참인 조건만 실행 | 혈압 등급 분류, 체중 등급 판정 |
| **독립 조건** (if-if-if) | 각각 별도로 확인할 때 | 모든 조건을 개별 확인 | 여러 증상 체크, 알레르기 확인 |
| **중첩 조건** (if 안의 if) | 단계별 확인이 필요할 때 | 조건을 만족해야 다음 단계 진행 | 응급도 평가 → 진료과 배정 |

````python
# 통합 환자 데이터
patient = {
    "name": "김환자",
    "age": 45,
    "symptoms": ["발열", "기침", "두통"],
    "insurance": True,
    "emergency_level": "응급",
    "department": None,
    "doctor": None
}

patient
````

### 독립 조건 - if - if -if

````python
# 1. 독립 조건: 각 증상별 개별 처방 (모두 실행 가능)
print("=== 증상별 개별 처방 ===")
if "발열" in patient["symptoms"]:
    print("해열제를 처방합니다.")

if "기침" in patient["symptoms"]:
    print("기침약을 처방합니다.")

if "두통" in patient["symptoms"]:
    print("진통제를 처방합니다.")
````

### 다중 조건 if-elif-else

````python
# 2. 다중 조건: 나이에 따른 진료과 배정 (하나만 선택)
print("=== 나이별 진료과 배정 ===")
if patient["age"] < 18:
    patient["department"] = "소아과"
elif patient["age"] < 65:
    patient["department"] = "내과"
else:
    patient["department"] = "노인의학과"

print(f"배정 진료과: {patient['department']}")
````

### 중첩 조건 - if 안에 if

````python
# 3. 중첩 조건: 단계별 진료 절차
print("=== 응급 환자 처리 절차 ===")

# 1단계: 응급도 확인
if patient["emergency_level"] == "응급":
    print("응급 환자입니다.")

    # 2단계: 보험 확인
    if patient["insurance"]:
        print("보험 적용으로 즉시 치료 시작합니다.")

        # 3단계: 진료과에 따른 담당의 배정
        if patient["department"] == "내과":
            patient["doctor"] = "김내과"
            print(f"담당의: {patient['doctor']} 배정완료")
        elif patient["department"] == "소아과":
            patient["doctor"] = "이소아"
            print(f"담당의: {patient['doctor']} 배정완료")
        else:
            print("담당의 배정 대기 중입니다.")

    else:
        print("보험 미적용이지만 응급이므로 치료 후 수납하겠습니다.")

else:
    print("일반 환자입니다. 접수 후 대기해주세요.")
````

## 💊 실전 예제: 환자 입원 승인 시스템



````python
patient = {
    "name": "이환자",  # 환자 이름
    "temperature": 37.2, # 체온 (섭씨)
    "bed_preference": "일반실",  # 희망 병실 유형
    "insurance_type": "건강보험",  # 보험 종류
    "emergency_contact": "010-1234-5678"  # 응급 연락처
}

# 병원 상황 (남은 병상 수)
hospital_status = {
    "일반실": 2,
    "특실": 5,
    "중환자실": 0
}
````

````python
print("=== 입원 가능 여부 판정 ===")

# 1단계: 바이탈 사인 확인
if patient["temperature"] < 38.0:
    print("체온 정상 범위입니다.")

    # 2단계: 병상 확인
    if hospital_status[patient["bed_preference"]] > 0:
        print(f"✅ {patient['bed_preference']} 입원 가능합니다.")

        # 3단계: 보험 확인
        if patient["insurance_type"] == "건강보험":
            print("건강보험 적용으로 입원 수속을 진행합니다.")
        else:
            print("실비보험 확인 후 입원 수속을 진행합니다.")

    else:
        print(f"❌ {patient['bed_preference']}에 여유 병상이 없습니다.")
        print("다른 병실을 확인하거나 대기 등록하시겠습니까?")

else:
    print("⚠️ 발열로 인해 격리병실 배정이 필요합니다.")
````

# 4. 반복문

반복문은 **특정 조건에 따라 똑같은 기능을 반복해서 수행**하는 구문입니다.   
의료 현장에서는 여러 환자의 데이터를 처리하거나, 검사 결과를 순차적으로 분석할 때 반복문이 필수적입니다.

<aside>

💡 **반복문 리딩 스킬**

반복문을 **"각각에 대해"**로 읽으면 코드 이해가 쉬워집니다!

</aside>



## 🔁 for문 - 정해진 횟수만큼

for문은 **정해진 범위나 리스트의 요소들을 순회하며 하나씩 처리**할 때 사용합니다.

````python
# 환자 명단 순회
patient_names = ["김환자", "이환자", "박환자", "최환자"]  # 오늘 진료 환자 명단

for patient_name in patient_names:
    print(f"{patient_name}님, 진료실로 들어오세요!")
````

반복문이 없다면

````python
# ---- loop 1 ----#
print(patient_names[0])

# ---- loop 1 ----#
print(patient_names[1])

# ---- loop 1 ----#
print(patient_names[2])

# ---- loop 1 ----#
print(patient_names[3])
````

## 🔍 유용한 반복문 함수들

| 함수 | 용도 | 반환값 | 예시 |
| --- | --- | --- | --- |
| `range()` | 연속된 숫자 생성 | 숫자 시퀀스 | `range(5)` → 0,1,2,3,4 |
| `enumerate()` | 인덱스와 값 함께 | (인덱스, 값) 튜플 | `enumerate(['a','b'])` → (0,'a'), (1,'b') |
| `zip()` | 여러 리스트 동시 순회 | 각 리스트의 요소들을 묶은 튜플 | `zip([1,2], ['a','b'])` → (1,'a'), (2,'b') |



### enumerate() - 인덱스와 값을 함께

`enumerate()`는 **인덱스(순서)와 값을 함께** 가져올 때 사용합니다.



````python
# 기본 사용법 - 환자 대기 순번
patients = ["김환자", "이환자", "박환자", "최환자"]  # 대기 환자 명단

for index, patient_name in enumerate(patients):
    print(f"{index + 1}번째 대기: {patient_name}")
````

## ♾️ while문 - 조건이 맞을 때까지

while문은 **조건이 참인 동안 계속 반복**합니다.


````python
# 기본 while문
count = 1

while count <= 5:
    print(f"현재 숫자: {count}")
    count += 1  # count = count + 1 과 같음

print("카운트 완료!")
````

활용하기

````python
# 혈압 안정화 대기
systolic_pressure = 160  # 현재 수축기 혈압
target_pressure = 140  # 목표 혈압
hour_count = 0  # 경과 시간

print("=== 혈압 안정화 치료 ===")
while systolic_pressure > target_pressure:
    hour_count += 1
    systolic_pressure -= 5  # 시간당 5씩 감소 가정
    print(f"{hour_count}시간 후: {systolic_pressure} mmHg")

print(f"혈압 안정화 완료! (총 {hour_count}시간 소요)")
````

## 🛑 break와 continue

- **break**: 반복문을 완전히 종료
- **continue**: 현재 반복을 건너뛰고 다음 반복으로


````python
# break - 특정 조건에서 반복 종료
for i in range(1, 11):
    if i == 5:
        print("5를 만났으니 종료!")
        break  # 반복문 완전 종료
    print(f"숫자: {i}")
````

````python
# continue - 특정 조건에서 건너뛰기
for i in range(1, 6):
    if i == 3:
        print("3은 건너뛰기!")
        continue  # 3일 때만 아래 코드 실행 안함
    print(f"숫자: {i}")
````

활용하기

````python
# 통합 환자 데이터
patient_data = {
    "김환자": {"systolic": 120, "allergic": False},  # 수축기 혈압, 알레르기 여부
    "이환자": {"systolic": 145, "allergic": True},
    "박환자": {"systolic": 160, "allergic": False},
    "최환자": {"systolic": 135, "allergic": True},
    "정환자": {"systolic": 180, "allergic": False}
}
````

````python
# break - 위험 혈압 환자 첫 발견시 즉시 응급 처치
danger_threshold = 170  # 위험 혈압 기준값

print("=== 응급 혈압 환자 검색 ===")
for patient_name, data in patient_data.items():
    systolic = data["systolic"]  # 수축기 혈압
    print(f"{patient_name} 혈압 확인: {systolic} mmHg")

    if systolic >= danger_threshold:
        print(f"🚨 위험! {patient_name}의 혈압이 {systolic} mmHg입니다!")
        print("즉시 응급 처치를 시작합니다.")
        break  # 위험 환자 발견시 즉시 중단

    print(f"{patient_name} 정상 범위")

print("\n응급 혈압 환자 검색을 종료합니다.")
````

````python
print("=== 페니실린 처방 가능 환자 ===")
# continue 예제 - 알레르기 환자 제외하고 처방
for patient_name, data in patient_data.items():
    if data["allergic"]:  # 알레르기 여부 확인
        print(f"{patient_name}: 알레르기로 처방 제외")
        continue  # 알레르기 환자는 건너뛰기

    systolic = data["systolic"]
    print(f"{patient_name}: 페니실린 처방 가능 (혈압: {systolic} mmHg)")
````

## 💊 실전 예제: 환자 검사 결과 분석 시스템



````python
# 환자 검사 결과 데이터
patient_results = [
    {"name": "김환자", "age": 65, "systolic": 145, "blood_sugar": 110},  # 환자별 검사 결과
    {"name": "이환자", "age": 45, "systolic": 130, "blood_sugar": 95},
    {"name": "박환자", "age": 32, "systolic": 120, "blood_sugar": 85},
    {"name": "최환자", "age": 70, "systolic": 160, "blood_sugar": 200}
]
````

````python
# 1. 모든 환자 기본 정보 출력
for patient in patient_results:
    name = patient["name"]  # 환자 이름
    age = patient["age"]  # 환자 나이
    systolic = patient["systolic"]  # 수축기 혈압
    blood_sugar = patient["blood_sugar"]  # 혈당 수치
    print(f"{name}({age}세): 혈압 {systolic}mmHg, 혈당 {blood_sugar}mg/dL")
````

````python
# 2. 고위험군 환자 찾기
high_risk_patients = []  # 고위험군 환자 이름 목록

for patient in patient_results:
    name = patient["name"]
    age = patient["age"]
    systolic = patient["systolic"]
    blood_sugar = patient["blood_sugar"]

    # 위험 요소 확인
    is_hypertension = systolic >= 140  # 고혈압 여부
    is_diabetes = blood_sugar >= 140  # 당뇨 여부
    is_elderly = age >= 65  # 고령자 여부

    # 2개 이상 조건 만족시 고위험군
    risk_factors = [is_hypertension, is_diabetes, is_elderly]
    if sum(risk_factors) >= 2:
        high_risk_patients.append(name)

# 고위험군 환자 순번 출력
if high_risk_patients:
    for rank, patient_name in enumerate(high_risk_patients, 1):
        print(f"{rank}순위 고위험군: {patient_name}")
else:
    print("고위험군 환자가 없습니다.")
````

````python
# 3. 검사 항목별 평균 계산 (리스트 컴프리헨션 활용)
systolic_values = [patient["systolic"] for patient in patient_results]  # 모든 혈압 수치
blood_sugar_values = [patient["blood_sugar"] for patient in patient_results]  # 모든 혈당 수치

avg_systolic = sum(systolic_values) / len(systolic_values)  # 평균 혈압
avg_blood_sugar = sum(blood_sugar_values) / len(blood_sugar_values)  # 평균 혈당

print(f"평균 수축기 혈압: {avg_systolic:.1f} mmHg")
print(f"평균 혈당: {avg_blood_sugar:.1f} mg/dL")
````

````python
# 4. 정상 범위 환자 비율
normal_bp_patients = [patient["name"] for patient in patient_results if patient["systolic"] < 140]
normal_bs_patients = [patient["name"] for patient in patient_results if patient["blood_sugar"] < 140]

print(f"\n정상 혈압 환자: {len(normal_bp_patients)}명 ({normal_bp_patients})")
print(f"정상 혈당 환자: {len(normal_bs_patients)}명 ({normal_bs_patients})")
````
