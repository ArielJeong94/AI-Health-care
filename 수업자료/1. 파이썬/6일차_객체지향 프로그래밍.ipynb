## 🎯 객체지향 프로그래밍이란

**객체지향 프로그래밍(OOP)은** `데이터`를 어떻게 구조화하는지 정의하고 해당 데이터를 수정할 수 있는 `함수`를 제공하는 프로그래밍 방식입니다.

- **프로그램을 실제 세상의 객체처럼 구현**하는 방식
- **데이터와 해당 데이터를 처리하는 코드를 하나로 묶어 관리**
- 예시: 환자라는 객체는 이름, 나이, 병력(데이터)과 진료받기, 처방받기(기능)을 가짐

### 의료 현장에서의 객체지향 필요성

의료 현장에서는 **환자, 의사, 병실, 의료기록** 등 다양한 개체들이 복잡하게 상호작용합니다.   
함수만으로는 이러한 복잡성을 효과적으로 관리하기 어렵습니다.


````python
# 함수형 접근의 한계
def create_patient(name, age, condition):
    return {"name": name, "age": age, "condition": condition}

def update_condition(patient, new_condition):
    patient["condition"] = new_condition
    return patient

def get_patient_info(patient):
    return f"환자: {patient['name']}, 상태: {patient['condition']}"

# 데이터와 기능이 분리되어 유지보수가 어려움
patient = create_patient("김환자", 45, "고혈압")
patient = update_condition(patient, "당뇨병")
print(get_patient_info(patient))
````

### 객체지향의 주요 특징


#### 1. 캡슐화 (Encapsulation)
**환자의 개인정보와 의료기록을 하나로 묶어 관리하고, 권한이 있는 사람만 접근할 수 있도록 보호**


#### 2. 상속 (Inheritance)  
**일반 직원 → 의사, 간호사로 확장하여 공통 특성은 재사용하고 고유 특성만 추가**


#### 3. 다형성 (Polymorphism)
**환자마다 다른 치료법이지만 '치료하기'라는 같은 인터페이스 사용**


#### 4. 추상화 (Abstraction)
**복잡한 의료 과정을 핵심 기능만 추출하여 단순하게 표현**

### 클래스 변수 vs 인스턴스 변수

**클래스 변수는 모든 인스턴스가 공유하는 변수이고, 인스턴스 변수는 각 객체마다 독립적으로 가지는 변수입니다.**

````python
class Hospital:
    # 클래스 변수 - 모든 병원이 공유하는 정보
    country = "대한민국"
    total_hospitals = 0  # 전체 병원 수

    def __init__(self, name, location):
        # 인스턴스 변수 - 각 병원의 고유 정보
        self.name = name
        self.location = location
        self.patient_count = 0

        # 병원이 생성될 때마다 전체 병원 수 증가
        Hospital.total_hospitals += 1 # 클래스 변수 수정
````

````python
# 사용 예시
print(f"병원 생성 전 총 병원 수: {Hospital.total_hospitals}")  # 0

# 병원 인스턴스 생성
hospital1 = Hospital("서울대병원", "서울")
hospital2 = Hospital("연세대병원", "서울")

# 클래스 변수는 모든 인스턴스가 공유
print(f"병원 생성 후 총 병원 수: {hospital1.total_hospitals}")   # 2
print(f"병원 생성 후 총 병원 수: {hospital2.total_hospitals}")   # 2
print(f"병원 생성 후 총 병원 수: {Hospital.total_hospitals}")    # 2

print(hospital1.country)           # 대한민국
print(hospital2.country)           # 대한민국
print(Hospital.country)            # 대한민국

# 인스턴스 변수는 각각 독립적
print(hospital1.name)              # 서울대병원
print(hospital2.name)              # 연세대병원
````

## 🚀 생성자와 self


### 생성자 (__init__ 메서드)

**__init__ 메서드는** 객체가 생성될 때 자동으로 호출되는 특별한 메서드입니다.   
객체의 초기 상태를 설정하고 인스턴스 변수를 정의하는 역할을 합니다.


````python
class Patient:
    def __init__(self, name, age, patient_id):
        """환자 객체를 초기화합니다"""
        print(f"환자 {name} 등록 중...")
        self.name = name           # 환자명
        self.age = age             # 나이
        self.patient_id = patient_id  # 환자 ID
        self.medical_history = []   # 병력
        self.current_status = "대기중"  # 현재 상태
        print(f"환자 {name} 등록 완료!")

# 인스턴스 생성 시 __init__이 자동 호출됨
patient = Patient("김환자", 45, "P001")
````

### self 키워드

**self는** 현재 객체 자신을 가리키는 **대명사** 역할을 합니다. 인스턴스의 속성과 메서드에 접근할 때 사용합니다.  
클래스 안의 모든 인스턴스 메소드는 자동으로 첫 번째 인자를 self로 받습니다.


````python
class Doctor:
    def __init__(self, name, specialty):
        """의사 정보를 초기화합니다"""
        self.name = name           # 이 객체의 name
        self.specialty = specialty  # 이 객체의 specialty
        self.patient_list = []     # 이 객체의 patient_list

    def introduce(self):
        """의사의 자기소개를 합니다"""
        # self를 통해 자신의 속성에 접근
        return f"안녕하세요, {self.specialty} 전문의 {self.name}입니다."

    def add_patient(self, patient_name):
        """환자를 담당 환자 목록에 추가합니다"""
        # self를 통해 자신의 리스트에 추가
        self.patient_list.append(patient_name)
        print(f"{patient_name} 환자가 {self.name} 의사에게 배정되었습니다.")
````

````python
# 두개의 인스턴스 생성
doctor1 = Doctor("김의사", "내과")
doctor2 = Doctor("이의사", "외과")
````

````python
# 각각 자신의 정보로 소개
print(doctor1.introduce())
print(doctor2.introduce())
````

````python
# 각각 자신의 환자 목록에 추가
doctor1.add_patient("환자A")
doctor2.add_patient("환자B")
````

````python
print(doctor1.patient_list)
print(doctor2.patient_list)
````

## 🛠️ 메서드 종류


### 인스턴스 메서드

**인스턴스 메서드는** 특정 객체가 수행할 수 있는 동작을 정의합니다.  
첫 번째 매개변수로 항상 `self`를 받아 객체의 속성에 접근하고 수정하여 객체가 수행할 수 있는 기능을 제공하는 역할을 합니다.

````python
class Patient:
    def __init__(self, name, age, condition):
        """환자 정보를 초기화합니다"""
        self.name = name          # 환자명
        self.age = age            # 나이
        self.condition = condition  # 진단명

    # 인스턴스 메서드
    def update_condition(self, new_condition):
        """환자의 진단명을 업데이트합니다"""
        old_condition = self.condition
        self.condition = new_condition
        print(f"{self.name} 환자: {old_condition} → {self.condition}")

    # 인스턴스 메서드
    def get_basic_info(self):
        """환자의 기본 정보를 반환합니다"""
        return f"{self.name}({self.age}세) - {self.condition}"
````

````python
# 환자 인스턴스 생성
patient = Patient("김환자", 45, "고혈압")

# 메서드 호출
print(patient.get_basic_info())
patient.update_condition("당뇨병")
print(patient.get_basic_info())
````

### 메서드 간 상호 호출

**메서드 안에서 다른 메서드를 호출**하여 코드 재사용성을 높일 수 있습니다.


````python
class Doctor:
    def __init__(self, name, specialty):
        """의사 정보를 초기화합니다"""
        self.name = name              # 의사명
        self.specialty = specialty    # 전문과목
        self.patient_count = 0        # 진료 환자 수

    def get_doctor_info(self):
        """의사 기본 정보를 반환합니다"""
        return f"{self.specialty} {self.name} 의사"

    def treat_patient(self, patient_name):
        """환자를 진료합니다 (다른 메서드 호출)"""
        # 다른 메서드를 호출하여 의사 정보 가져오기
        doctor_info = self.get_doctor_info()

        self.patient_count += 1
        return f"{doctor_info}가 {patient_name} 환자를 진료했습니다. (총 {self.patient_count}명)"
````

````python
# 환자 인스턴스 생성
doctor = Doctor("김의사", "내과")

# 메서드 호출
print(doctor.treat_patient("환자A"))
print(doctor.treat_patient("환자B"))
````

### __str__ 메서드

**__str__ 메서드는** 객체를 문자열로 표현하는 방법을 정의하는 특별한 메서드입니다.  
`print()` 함수로 객체를 출력하거나 `str()` 함수를 사용할 때 자동으로 호출됩니다.


````python
# __str__ 메서드 없는 클래스
class PatientBasic:
    def __init__(self, name, age, patient_id):
        """환자 정보를 초기화합니다"""
        self.name = name              # 환자명
        self.age = age                # 나이
        self.patient_id = patient_id  # 환자번호

# __str__ 메서드 있는 클래스
class PatientStr:
    def __init__(self, name, age, patient_id):
        """환자 정보를 초기화합니다"""
        self.name = name              # 환자명
        self.age = age                # 나이
        self.patient_id = patient_id  # 환자번호

    def __str__(self):
        """환자 객체를 문자열로 표현합니다"""
        return f"환자 {self.patient_id}: {self.name}({self.age}세)"
````

````python
# 비교
patient1 = PatientBasic("김환자", 45, "P001")
patient2 = PatientStr("이환자", 30, "P001")

print(f"PatientBasic: {patient1}")
print(f"PatientStr: {patient2}")
````

## 🔄 상속 (Inheritance)

**상속은** 부모 클래스의 속성과 메서드를 자식 클래스가 물려받는 기능입니다.  
코드 재사용성을 높이고 계층적 구조로 체계적인 클래스 설계를 가능하게 합니다.


### 상속의 필요성과 장점

**코드 중복을 방지하고 체계적인 클래스 설계를 위해 상속이 필요합니다.**

의료진은 모두 **공통 속성**(이름, 근무부서)을 가지고 있지만, **고유 속성**(전문분야, 근무형태)은 직종마다 다릅니다.


````python
# 상속 없이 - 코드 중복 발생
class Doctor:
    def __init__(self, name, department, specialty):
        """의사 정보를 초기화합니다"""
        self.name = name          # 중복!
        self.department = department  # 중복!
        self.specialty = specialty

    def introduce(self):          # 중복!
        """자기소개를 합니다"""
        return f"안녕하세요, {self.department}의 {self.name}입니다."

class Nurse:
    def __init__(self, name, department, shift):
        """간호사 정보를 초기화합니다"""
        self.name = name          # 중복!
        self.department = department  # 중복!
        self.shift = shift

    def introduce(self):          # 중복!
        """자기소개를 합니다"""
        return f"안녕하세요, {self.department}의 {self.name}입니다."
````

### 부모 클래스와 자식 클래스

**부모 클래스는 공통 기능을 정의하고, 자식 클래스는 이를 상속받아 고유 기능을 추가합니다.**


````python
# 부모 클래스 (기본 의료진)
class MedicalStaff:
    def __init__(self, name, department):
        """의료진 기본 정보를 초기화합니다"""
        self.name = name              # 이름
        self.department = department  # 근무부서

    def introduce(self):
        """자기소개를 합니다"""
        return f"안녕하세요, {self.department}의 {self.name}입니다."

    def work(self):
        """기본 업무를 수행합니다"""
        return f"{self.name}이(가) 업무를 수행합니다."
````

````python

# 자식 클래스 1: 의사
class Doctor(MedicalStaff):  # 상속 문법: class 자식(부모)
    def __init__(self, name, department, specialty):
        """의사 정보를 초기화합니다"""
        # 부모 클래스의 __init__ 호출
        super().__init__(name, department)

        # 의사만의 고유 속성 추가
        self.specialty = specialty    # 전문분야

    def diagnose(self, patient_name):
        """환자를 진단합니다"""
        return f"🔴 {self.name} 의사가 {patient_name} 환자를 진단했습니다."

# 자식 클래스 2: 간호사
class Nurse(MedicalStaff):  # MedicalStaff 상속
    def __init__(self, name, department, shift):
        """간호사 정보를 초기화합니다"""
        # 부모 클래스의 __init__ 호출
        super().__init__(name, department)

        # 간호사만의 고유 속성 추가
        self.shift = shift            # 근무시간

    def care_patient(self, patient_name):
        """환자를 돌봅니다"""
        return f"🟣 {self.name} 간호사가 {patient_name} 환자를 돌보고 있습니다."
````

````python

# 인스턴스 생성
doctor = Doctor("김의사", "내과", "심장내과")
nurse = Nurse("이간호사", "외과", "주간")
````

````python
# 부모 클래스의 메서드 사용 (상속받음)
print(doctor.introduce())
print(nurse.introduce())
````

````python
# 자식 클래스의 고유 메서드 사용
print(doctor.diagnose("환자A"))
print(nurse.care_patient("환자B"))
````

### super() 함수

**super() 함수는 자식 클래스에서 부모 클래스의 메서드를 호출할 때 사용하는 내장 함수입니다.**

이를 통해 부모 클래스의 기능을 유지하면서 새로운 기능을 추가할 수 있습니다.


````python

class Patient:
    def __init__(self, name, age):
        """환자 기본 정보를 초기화합니다"""
        self.name = name      # 환자명
        self.age = age        # 나이
        self.status = "대기중"  # 상태
        print(f"환자 {self.name} 등록 완료")

    def get_info(self):
        """환자 기본 정보를 반환합니다"""
        return f"환자: {self.name}({self.age}세)"

class InPatient(Patient):
    def __init__(self, name, age, room_number):
        """입원 환자 정보를 초기화합니다"""
        # 1. 부모 클래스의 __init__ 먼저 호출
        super().__init__(name, age)

        # 2. 입원 환자만의 추가 정보
        self.room_number = room_number    # 병실번호
        print(f"{self.name} 환자가 {room_number}호에 입원했습니다.")

    def get_info(self):
        """입원 환자 정보를 반환합니다 (기능 확장)"""
        # 1. 부모 클래스의 기본 정보 먼저 가져오기
        basic_info = super().get_info()

        # 2. 입원 환자만의 추가 정보 포함
        return f"{basic_info}, 병실: {self.room_number}호"

````

````python
# 인스턴스 생성
patient = Patient("김환자", 45)
inpatient = InPatient("박환자", 30, "301")
````

````python
# 인스턴스 메서드 호출
print(patient.get_info())
print(inpatient.get_info())
````

### 심화 - 계층적 클래스 구조 설계

**여러 단계의 상속을 통해 더욱 세분화된 클래스 계층을 만들 수 있습니다.**


````python
# 3단계 상속 구조
class MedicalStaff:
    """의료진 기본 클래스"""
    def __init__(self, name):
        self.name = name      # 이름

    def work(self):
        return f"{self.name}이(가) 업무를 수행합니다."

class Doctor(MedicalStaff):
    """의사 클래스"""
    def __init__(self, name, specialty):
        super().__init__(name)
        self.specialty = specialty    # 전문분야

class Specialist(Doctor):
    """전문의 클래스 (의사를 상속)"""
    def __init__(self, name, specialty, subspecialty):
        super().__init__(name, specialty)
        self.subspecialty = subspecialty  # 세부전문분야

    def research(self, topic):
        """연구를 수행합니다"""
        return f"{self.name} 전문의가 '{topic}' 연구를 진행합니다."
````

````python
# 인스턴스 생성
specialist = Specialist("박전문의", "내과", "심장내과")

# 인스턴스 메서드 호출
print(specialist.work())                    # MedicalStaff에서 상속
print(specialist.research("부정맥 치료"))    # Specialist 고유 메서드
````

## 🔀 메서드 오버라이딩

**메서드 오버라이딩은** 부모 클래스의 메서드를 자식 클래스에서 다시 정의하여 다른 동작을 하게 만드는 기법입니다.  


### 오버라이딩의 개념

**같은 이름의 메서드를 자식 클래스에서 다시 정의하여 부모 클래스와는 다른 동작을 수행하도록 만듭니다.**

파일에서 덮어쓰기와 같은 개념입니다.


````python
# 부모 클래스 (기본 의료진)
class MedicalStaff:
    def __init__(self, name, department):
        """의료진 기본 정보를 초기화합니다"""
        self.name = name              # 이름
        self.department = department  # 부서

    def introduce(self):
        """기본 자기소개"""
        return f"안녕하세요, {self.department}의 {self.name}입니다."

    def work(self):
        """기본 업무"""
        return f"{self.name}이(가) 일반적인 업무를 수행합니다."
````

````python
# 자식 클래스 1: 의사
class Doctor(MedicalStaff):
    def __init__(self, name, department, specialty):
        """의사 정보를 초기화합니다"""
        super().__init__(name, department)
        self.specialty = specialty    # 전문분야

    def introduce(self):  # 부모 메서드를 오버라이딩
        """의사의 자기소개 (다른 내용으로 재정의)"""
        return f"안녕하세요, {self.specialty} 전문의 {self.name}입니다."

    def work(self):  # 부모 메서드를 오버라이딩
        """의사의 업무 (다른 내용으로 재정의)"""
        return f"{self.name} 의사가 환자를 진료하고 있습니다."

# 자식 클래스 2: 간호사
class Nurse(MedicalStaff):
    def __init__(self, name, department, shift):
        """간호사 정보를 초기화합니다"""
        super().__init__(name, department)
        self.shift = shift            # 근무시간

    def introduce(self):  # 부모 메서드를 오버라이딩
        """간호사의 자기소개 (다른 내용으로 재정의)"""
        return f"안녕하세요, {self.shift} 근무 간호사 {self.name}입니다."

    def work(self):  # 부모 메서드를 오버라이딩
        """간호사의 업무 (다른 내용으로 재정의)"""
        return f"{self.name} 간호사가 환자를 돌보고 있습니다."
````

````python
# 인스턴스 생성

staff_list = [
    MedicalStaff("김직원", "행정부"),
    Doctor("이의사", "내과", "심장내과"),
    Nurse("박간호사", "외과", "주간")
]

# 같은 메서드명이지만 다른 동작 (다형성)
for staff in staff_list:
    print(staff.introduce())  # 각자의 오버라이딩된 메서드 호출
    print(staff.work())
    print("-" * 30)

````

### 완전 재정의 vs 기능 확장

**오버라이딩에는 완전히 새로운 기능으로 바꾸는 방법과 기존 기능을 확장하는 방법이 있습니다.**


````python
# 부모 클래스 (기본 의료기록)
class MedicalRecord:
    def __init__(self, patient_name, doctor_name):
        """의료기록 기본 정보를 초기화합니다"""
        self.patient_name = patient_name    # 환자명
        self.doctor_name = doctor_name      # 의사명
        self.records = []                   # 기록 목록

    def add_record(self, content):
        """기록을 추가합니다"""
        self.records.append(content)
        print(f"기록 추가: {content}")

    def get_summary(self):
        """기본 요약 정보를 반환합니다"""
        return f"환자: {self.patient_name}, 담당의: {self.doctor_name}, 기록 수: {len(self.records)}개"

````

````python

# 자식 클래스 (수술 기록)
class SurgeryRecord(MedicalRecord):
    def __init__(self, patient_name, doctor_name, surgery_type):
        """수술 기록 정보를 초기화합니다"""
        super().__init__(patient_name, doctor_name)
        self.surgery_type = surgery_type    # 수술명

    def add_record(self, content, phase="일반"):
        """수술 기록 추가 (기능 확장 방식)"""
        # 1. 부모 클래스의 기본 기록 추가 기능 활용
        super().add_record(content)

        # 2. 수술 기록만의 추가 처리
        print(f"수술 단계: {phase}")

    def get_summary(self):
        """수술 요약 정보 반환 (완전 재정의 방식)"""
        # 부모 클래스의 기본 요약은 사용하지 않고 완전히 새로 작성
        return f"수술: {self.surgery_type}, 환자: {self.patient_name}, 집도의: {self.doctor_name}"

````

````python
# 인스턴스 생성
surgery = SurgeryRecord("김환자", "박외과의", "충수절제술")

# 기능 확장된 메서드 사용
surgery.add_record("마취 시작", "수술 전")

````

````python
# 완전 재정의된 메서드 사용
print(surgery.get_summary())
````
