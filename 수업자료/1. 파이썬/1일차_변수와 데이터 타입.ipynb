## 💡 변수란?

변수는 **데이터를 저장하는 상자**입니다. 상자에 이름표를 붙여서 나중에 쉽게 찾아 쓸 수 있습니다.

````python
# 변수 정의
name = "김개발"      # 문자열 저장
age = 25            # 숫자 저장
height = 175.5      # 소수점 숫자 저장
is_student = True   # 참/거짓 저장

# 변수 사용
print(name)
print(age)
print(height)
print(is_student)
````

## 🔍 데이터 타입 확인하기

파이썬에서는 **type()** 함수와 **isinstance()** 함수를 사용해서 데이터의 타입을 확인할 수 있습니다.

- `type(변수)`: 변수의 정확한 타입을 반환
- `isinstance(변수, 타입)`: 변수가 특정 타입인지 True/False로 반환

````python
name = "홍길동"
age = 25
height = 175.5
is_diabetic = False
````

````python
# type() 함수로 데이터 타입 확인

print(type(name))
print(type(age))
print(type(height))
print(type(is_diabetic))
````

````python
# 특정 타입인지 확인
print(type(name) == str)
print(type(age) == int)
print(type(height) == float)
print(type(is_diabetic) == bool)
````

````python
# isinstance() 함수로 특정 타입인지 확인
print(isinstance(age, int))
print(isinstance(name, str))
print(isinstance(height, int))
print(isinstance(is_diabetic, bool))
````

## 🔢 연산자 활용

### 🧮 산술 연산자

````python
a = 10
b = 3

print(f"{a} + {b} = {a + b}")    # 10 + 3 = 13 (더하기)
print(f"{a} - {b} = {a - b}")    # 10 - 3 = 7  (빼기)
print(f"{a} * {b} = {a * b}")    # 10 * 3 = 30 (곱하기)
print(f"{a} / {b} = {a / b}")    # 10 / 3 = 3.333... (나누기)
print(f"{a} // {b} = {a // b}")  # 10 // 3 = 3 (몫)
print(f"{a} % {b} = {a % b}")    # 10 % 3 = 1 (나머지)
print(f"{a} ** {b} = {a ** b}")  # 10 ** 3 = 1000 (거듭제곱)
````

#### 활용 예시


````python
# 약물 용량 계산
weight = 70      # 환자 체중 (kg)
dose_per_kg = 5  # kg당 용량 (mg)

print(f"총 투여량: {weight * dose_per_kg}mg")           # 총 투여량: 350mg
print(f"1회 용량: {(weight * dose_per_kg) / 3}mg")      # 1회 용량: 116.67mg
print(f"용량 차이: {(weight * dose_per_kg) - 300}mg")   # 용량 차이: 50mg
print(f"체중 제곱: {weight ** 2}")                      # 체중 제곱: 4900
print(f"남은 알약: {21 % 7}개")                        # 남은 알약: 0개
print(f"몇 주치: {21 // 7}주")                         # 몇 주치: 3주
````

### ⚖️ 비교 연산자

````python
# 비교 연산자의 결과값은 bool타입

x = 5
y = 3

print(f"{x} > {y} = {x > y}")    # 5 > 3 = True
print(f"{x} < {y} = {x < y}")    # 5 < 3 = False
print(f"{x} >= {y} = {x >= y}")  # 5 >= 3 = True
print(f"{x} <= {y} = {x <= y}")  # 5 <= 3 = False
print(f"{x} == {y} = {x == y}")  # 5 == 3 = False (같다)
print(f"{x} != {y} = {x != y}")  # 5 != 3 = True (같지않다)
````

#### 활용 예시


````python
# 혈압 수치 비교
systolic = 140 # 환자의 수축기 혈압
normal_systolic = 120 # 정상 수축기 혈압 기준값

print(f"수축기 혈압 {systolic} > 정상 {normal_systolic}: {systolic > normal_systolic}")
print(f"혈압 정상여부: {systolic <= normal_systolic}")
print(f"고혈압 진단: {systolic >= 140}")
print(f"정상 혈압: {systolic == normal_systolic}")
print(f"혈압 이상: {systolic != normal_systolic}")
````

### 🔗 논리 연산자

논리 연산자는 **여러 조건을 조합**할 때 사용합니다. 의료 진단에서 여러 증상을 종합 판단하는 것과 비슷합니다.


#### 🤝 and 연산자 (그리고)

**모든 조건이 참일 때만** 전체가 참이 됩니다.


````python
# 환자 정보
age = 65            # 나이
has_fever = True    # 발열 여부
has_cough = False    # 기침 여부

# 코로나 의심 조건: 발열이 있고 기침도 있을 때
covid_suspect = has_fever and has_cough
print(f"코로나 의심: {covid_suspect}")

# 고위험군 조건: 65세 이상이고 발열이 있을 때
high_risk = (age >= 65) and has_fever
print(f"고위험군: {high_risk}")
````

#### 🔄 or 연산자 (또는)

**조건 중 하나라도 참이면** 전체가 참이 됩니다.


````python
# 응급실 내원 기준
chest_pain = True     # 흉통 여부
difficulty_breathing = False  # 호흡곤란 여부
severe_headache = False      # 심한 두통 여부

# 응급실 내원: 흉통, 호흡곤란, 심한 두통 중 하나라도 있으면
emergency_visit = chest_pain or difficulty_breathing or severe_headache
print(f"응급실 내원 필요: {emergency_visit}")

# 처방전 필요: 발열이나 기침이 있으면
needs_prescription = has_fever or has_cough
print(f"처방전 필요: {needs_prescription}")
````

#### 🚫 not 연산자 (아니다)

**조건을 반대로** 만들어줍니다.


```python
# 환자 상태
has_symptoms = False    # 증상 여부
is_pregnant = True      # 임신 여부

# 건강한 상태: 증상이 없을 때
is_healthy = not has_symptoms
print(f"건강 상태: {is_healthy}")  # True

# 일반 약물 처방 가능: 임신이 아닐 때
can_prescribe_general_medicine = not is_pregnant
print(f"일반 약물 처방 가능: {can_prescribe_general_medicine}")  # False
```


#### 활용 예시


````python
# 환자 정보
patient_age = 70
has_diabetes = True      # 당뇨병 여부
blood_pressure = 160     # 수축기 혈압
has_heart_disease = False # 심장병 여부

# 복합 조건 판정
# 고혈압 진단: 혈압이 140 이상
has_hypertension = blood_pressure >= 140

# 심혈관 위험군: (당뇨병이 있거나 심장병이 있고) 고혈압인 경우
cardiovascular_risk = (has_diabetes or has_heart_disease) and has_hypertension
print(f"심혈관 위험군: {cardiovascular_risk}")

# 정기검진 대상: 65세 이상이고 만성질환이 없는 경우
chronic_disease = has_diabetes or has_heart_disease or has_hypertension
regular_checkup_needed = (patient_age >= 65) and (not chronic_disease)
print(f"정기검진 필요: {regular_checkup_needed}")
````

#### 논리 연산자 우선순위


````python
# 괄호를 사용해서 명확하게 표현하는 것이 좋습니다
age = 75
fever = True
cough = False

# 헷갈리는 표현
result1 = age > 65 and fever or cough  # 의도가 불분명

# 명확한 표현
result2 = (age > 65 and fever) or cough  # 의도가 명확
result3 = age > 65 and (fever or cough)  # 다른 의미

print(f"결과1: {result1}")
print(f"결과2: {result2}")
print(f"결과3: {result3}")
````
