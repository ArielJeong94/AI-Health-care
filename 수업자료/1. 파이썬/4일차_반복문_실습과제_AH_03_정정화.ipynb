# 실습과제: 환자 혈압 분류하기

병원에서 제공하는 환자 혈압 데이터를 활용하여 혈압 분류, 통계 분석, 고위험군 관리를 수행하는 프로그램을 작성해보세요.

## 요구사항

1. **혈압 분류 기준**
   - 정상: 수축기 < 120
   - 고혈압 전단계: 120 ≤ 수축기 < 140  
   - 1기 고혈압: 140 ≤ 수축기 < 160
   - 2기 고혈압: 수축기 ≥ 160

2. **기능 구현**
   - 각 환자별 혈압 분류 및 위험도 평가
   - **응급 처치가 필요한 환자(수축기 180 이상) 발견 시 즉시 감지 및 중단**
   - 혈압 등급별 통계 및 전체 환자 현황 분석

3. **출력 형식**
```
=== 개별 환자 혈압 분류 ===
김환자(65세): 145 mmHg - 1기 고혈압 (약물 치료 필요)
이환자(45세): 125 mmHg - 고혈압 전단계 (생활습관 개선)
박환자(32세): 115 mmHg - 정상 (정기 검진)
최환자(28세): 135 mmHg - 고혈압 전단계 (생활습관 개선)
정환자(70세): 190 mmHg - 2기 고혈압 (즉시 치료 필요)

=== 응급 처치 대상 ===
🚨 응급: 정환자 (190 mmHg) - 즉시 치료 필요

=== 혈압 등급별 통계 ===
정상: 2명 (40.0%)
고혈압 전단계: 1명 (20.0%)  
1기 고혈압: 1명 (20.0%)
2기 고혈압: 1명 (20.0%)

=== 전체 현황 요약 ===
총 환자 수: 5명
평균 혈압: 142.0 mmHg
고혈압 환자: 2명 (40.0%)
응급 처치 대상: 1명
```

### 힌트

- `if-elif-else`를 사용해 혈압 등급을 분류하세요
- `break`를 사용해 응급 환자 발견시 즉시 중단하세요
- 리스트 컴프리헨션을 사용해 조건별 환자를 필터링하세요
- `len()` 함수로 환자 수를 계산하세요
- `sum()` 함수로 평균 혈압을 계산하세요
- f-string을 사용해 결과를 깔끔하게 출력하세요


## 💻 구현해보세요

````python
# 기본 데이터 - 환자 혈압 데이터베이스
patients = [
    {"name": "김환자", "age": 65, "systolic": 145},    # 이름, 나이, 수축기 혈압(mmHg)
    {"name": "이환자", "age": 45, "systolic": 125},    # 이름, 나이, 수축기 혈압(mmHg)
    {"name": "박환자", "age": 32, "systolic": 115},    # 이름, 나이, 수축기 혈압(mmHg)
    {"name": "최환자", "age": 28, "systolic": 135},    # 이름, 나이, 수축기 혈압(mmHg)
    {"name": "정환자", "age": 70, "systolic": 190}     # 이름, 나이, 수축기 혈압(mmHg)
]
````

````python
# 여기에 코드를 작성하세요

# 1. 개별 환자 혈압 분류 (if-elif-else 활용)
emergency_patients = [] # 응급 환자 이름만 모으는 용으로 리스트 생성

print("=== 개별 환자 혈압 분류 ===")
for patient in patients:
  name, age, systolic = patient.values() # value값 언패킹해서 할당

  # 혈압 분류
  if systolic < 120:
    result = "정상"
    treat_way = "정기 검진"
  elif systolic < 140:
    result = "고혈압 전단계"
    treat_way = "생활습관 개선"
  elif systolic < 160:
    result = "1기 고혈압"
    treat_way = "약물 치료 필요"
  else:
    result = "2기 고혈압"
    treat_way = "즉시 치료 필요"

  print(f"{name}({age}세): {systolic} mmHg - {result} ({treat_way})")

# 2. 응급 처치 대상 즉시 감지 (break 활용)
  if systolic < 180:
    continue
  elif systolic >= 180:
    print("\n=== 응급 처치 대상 ===")
    print(f"🚨 응급: {name} ({systolic} mmHg) - 즉시 치료 필요")
    emergency_patients.append(name) # 응급 환자 이름 추가
    break

# 3. 혈압 등급별 통계 계산 (리스트 컴프리헨션 활용)
print("\n=== 혈압 등급별 통계 ===")

# 정상 환자 수
normal_count = len([patient for patient in patients if patient['systolic'] < 120])
# 고혈압 전단계 환자 수
pre_step_count = len([patient for patient in patients if patient['systolic'] >= 120 and patient['systolic'] < 140])
# 1기 고혈압 환자 수
first_step_count = len([patient for patient in patients if patient['systolic'] >= 140 and patient['systolic'] < 160])
# 2기 고혈압 환자 수
second_step_count = len([patient for patient in patients if patient['systolic'] >= 160])
# 전체 환자 수
total = len(patients)

print(f"정상: {normal_count}명 ({normal_count / total * 100}%)")
print(f"고혈압 전단계: {pre_step_count}명 ({pre_step_count/ total * 100}%)")
print(f"1기 고혈압: {first_step_count}명 ({first_step_count / total * 100}%)")
print(f"2기 고혈압: {second_step_count}명 ({second_step_count / total * 100}%)")

# 4. 전체 현황 요약 (내장 함수 활용)
print("\n=== 전체 현황 요약 ===")

# 혈압 리스트 생성
pressure_list = []
for patient in patients:
  name, age, systolic = patient.values()
  pressure_list.append(systolic)

print(f"총 환자 수: {total}명")
print(f"평균 혈압: {sum(pressure_list) / total} mmHg")
print(f"고혈압 환자: {first_step_count + second_step_count}명 ({(first_step_count + second_step_count) / total * 100}%)")
print(f"응급 처치 대상: {len(emergency_patients)}명")
````

````python
# @title 정답

# 환자 혈압 데이터베이스
patients = [
    {"name": "김환자", "age": 65, "systolic": 145},    # 이름, 나이, 수축기 혈압(mmHg)
    {"name": "이환자", "age": 45, "systolic": 125},    # 이름, 나이, 수축기 혈압(mmHg)
    {"name": "박환자", "age": 32, "systolic": 115},    # 이름, 나이, 수축기 혈압(mmHg)
    {"name": "최환자", "age": 28, "systolic": 135},    # 이름, 나이, 수축기 혈압(mmHg)
    {"name": "정환자", "age": 70, "systolic": 190}     # 이름, 나이, 수축기 혈압(mmHg)
]

# 1. 개별 환자 혈압 분류
print("=== 개별 환자 혈압 분류 ===")
for patient in patients:
    name, age, systolic = patient.values()    # 언패킹 (환자 이름, 환자 나이, 수축기 혈압(mmHg))

    # 혈압 분류 및 권고사항
    if systolic < 120:
        category = "정상"
        recommendation = "정기 검진"
    elif systolic < 140:
        category = "고혈압 전단계"
        recommendation = "생활습관 개선"
    elif systolic < 160:
        category = "1기 고혈압"
        recommendation = "약물 치료 필요"
    else:
        category = "2기 고혈압"
        recommendation = "즉시 치료 필요"

    print(f"{name}({age}세): {systolic} mmHg - {category} ({recommendation})")

# 2. 응급 처치 대상 즉시 감지
print("\n=== 응급 처치 대상 ===")
emergency_found = False

for patient in patients:
    name, age, systolic = patient.values()    # 언패킹 (환자 이름, 수축기 혈압(mmHg))

    if systolic >= 180:  # 응급 상황 기준
        print(f"🚨 응급: {name} ({systolic} mmHg) - 즉시 치료 필요")
        emergency_found = True
        break  # 응급 환자 발견시 즉시 처리

if not emergency_found:
    print("응급 처치가 필요한 환자가 없습니다.")

# 3. 혈압 등급별 통계 계산
print("\n=== 혈압 등급별 통계 ===")

# 각 등급별 환자 수 계산
normal_count = len([p for p in patients if p["systolic"] < 120])              # 정상 환자 수
prehyper_count = len([p for p in patients if 120 <= p["systolic"] < 140])     # 고혈압 전단계 환자 수
stage1_count = len([p for p in patients if 140 <= p["systolic"] < 160])       # 1기 고혈압 환자 수
stage2_count = len([p for p in patients if p["systolic"] >= 160])             # 2기 고혈압 환자 수

total_patients = len(patients)

# 비율 계산 및 출력
print(f"정상: {normal_count}명 ({normal_count/total_patients*100:.1f}%)")
print(f"고혈압 전단계: {prehyper_count}명 ({prehyper_count/total_patients*100:.1f}%)")
print(f"1기 고혈압: {stage1_count}명 ({stage1_count/total_patients*100:.1f}%)")
print(f"2기 고혈압: {stage2_count}명 ({stage2_count/total_patients*100:.1f}%)")

# 4. 전체 현황 요약
print("\n=== 전체 현황 요약 ===")

# 혈압 값들 추출
all_systolic = [patient["systolic"] for patient in patients]

# 통계 계산
total_patients = len(patients) # 총 환자 수
avg_systolic = sum(all_systolic) / len(all_systolic) # 평균 혈압
hypertension_patients = len([p for p in patients if p["systolic"] >= 140]) # 고혈압 환자 수
emergency_patients = len([p for p in patients if p["systolic"] >= 180]) # 응급 처치 대상 수

print(f"총 환자 수: {total_patients}명")
print(f"평균 혈압: {avg_systolic:.1f} mmHg")
print(f"고혈압 환자: {hypertension_patients}명 ({hypertension_patients/total_patients*100:.1f}%)")
print(f"응급 처치 대상: {emergency_patients}명")
````
