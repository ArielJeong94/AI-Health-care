```python
def cal_bmi(weight, height):
    height_m = height/100
    return int(weight/(height_m ** 2))

print(cal_bmi(70, 175))
```

```python
cal_bmi_2 = lambda weight, height: weight/((height/100)**2)
print(cal_bmi_2(70, 175))
```

```python
def classify_bmi(cal_bmi_2):
    if cal_bmi_2 <= 18.5:
        return("저체중")
    elif 18.5 < cal_bmi_2 <= 23:
        return("정상체중")
    elif 23 < cal_bmi_2 <= 25:
        return("과체중")
    else:
        return("비만")

print(classify_bmi(cal_bmi(70, 175)))
```

```python
class patient:
    def __init__(self, name, dept):
        self.name = name
        self.dept = dept

    def change_dept(self, new_dept):
        old_dept = self.dept
        self.dept = new_dept
        print(f"{self.name}님의 담당과가 {old_dept}에서 {self.dept}(으)로 변경되었습니다.")

patientList = []

for i, person in enumerate([{"name": "홍길동", "dept": "내분비내과"}, {"name": "심봉사", "dept": "안과"}]):
    patientList.append(patient(name = person["name"], dept = person["dept"]))

patientList[0].change_dept("신경외과")
patientList[1].change_dept("노년내과")
```
