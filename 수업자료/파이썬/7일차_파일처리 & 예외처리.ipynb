# 파일 입출력 & 예외처리

## 🗂️ 파일 입출력

**파일 입출력은** 프로그램에서 외부 파일에 데이터를 저장하거나 읽어오는 작업입니다.  
메모리에 있는 데이터는 프로그램이 종료되면 사라지지만, 파일에 저장하면 영구적으로 보존됩니다.


### 파일 입출력의 필요성

파일 입출력은 `프로그램`과 `외부 파일` 간의 데이터 교환을 통해 정보를 영구적으로 보존하고 공유할 수 있게 해줍니다.



### 의료 현장에서의 파일 활용

- **환자 진료기록**: 진단명, 처방 내역, 치료 과정을 파일로 기록
- **검사 결과**: 혈액검사, 영상검사 데이터를 파일에 저장
- **병원 운영**: 직원 정보, 일정 관리, 재고 현황을 파일로 관리
- **데이터 백업**: 중요한 의료 정보를 안전하게 보관


### open() 함수

**open() 함수를 사용하여 파일을 열고, 모드를 지정하여 어떤 작업을 할지 결정합니다.**

```python
# 기본 문법
file = open('파일명.txt', '모드', encoding='utf-8')
# 파일 작업 수행
file.close()  # 반드시 닫아야 함
```


### 파일 모드

파일 모드는 파일을 열 때 어떤 작업을 할지 결정하는 옵션입니다.

**주요 파일 모드**

- `'r'`: 읽기 모드 (기본값) - 기존 파일 내용을 읽음
- `'w'`: 쓰기 모드 - 새로운 내용으로 덮어쓰기 (기존 내용 삭제)
- `'a'`: 추가 모드 - 기존 내용 뒤에 새로운 내용 추가
- `'r+'`: 읽기/쓰기 모드 - 파일을 읽고 수정


### close() 메서드

**파일 작업이 끝나면 반드시 close() 메서드로 파일을 닫아야 합니다.**

파일을 닫지 않으면 메모리 누수가 발생하고 다른 프로그램에서 해당 파일에 접근하지 못할 수 있습니다.


````python
# 검사 결과 파일 생성
test_file = open('blood_test.txt', 'w', encoding='utf-8')
test_results = "혈당: 95mg/dL\n콜레스테롤: 180mg/dL\n혈압: 120/80mmHg"  # 검사 결과 데이터
test_file.write(test_results)
test_file.close()  # 파일 닫기 - 매우 중요!

print("검사 결과 파일이 저장되었습니다.")
````

## 🔒 with 문 - 권장

**with 문은** 파일을 자동으로 닫아주는 안전한 파일 처리 방법입니다.


### with 구문의 필요성

```python
# ❌ 위험한 방법
patient_file = open('patient_info.txt', 'w', encoding='utf-8')
# 오류 발생 시 파일이 닫히지 않을 수 있음!
patient_file.write("환자 정보")
patient_file.close()

# ✅ 안전한 방법
with open('patient_info.txt', 'w', encoding='utf-8') as file:
    file.write("환자 정보")
# 자동으로 파일이 닫힘
```


### 자동 파일 닫기 보장


````python
def save_patient_info(patient_id, name, diagnosis):
    """환자 정보를 파일에 안전하게 저장합니다"""
    filename = f"patient_{patient_id}.txt"

    with open(filename, 'w', encoding='utf-8') as file:
        file.write(f"환자ID: {patient_id}\n")      # 환자 식별번호
        file.write(f"환자명: {name}\n")            # 환자 이름
        file.write(f"진단명: {diagnosis}\n")       # 진단 결과

    print(f"{name} 환자의 기록이 저장되었습니다.")
````

````python

# 사용
save_patient_info("P001", "김환자", "고혈압")
````

## 📖 파일 읽기 - read()

**파일의 내용을 다양한 방법으로 읽을 수 있습니다.**


````python
# 환자 진료기록 파일 생성
with open('medical_records.txt', 'w', encoding='utf-8') as file:
    file.write("2024-03-15 김환자 고혈압 초진\n")
    file.write("2024-03-15 이환자 당뇨병 재진\n")
    file.write("2024-03-15 박환자 감기 초진\n")
````

````python
# 1. 전체 내용 읽기
with open('medical_records.txt', 'r', encoding='utf-8') as file:
    content = file.read()               # 전체 파일 내용
    print("=== 전체 진료기록 ===")
    print(content)
````

````python
# 2. 첫 번째 줄만 읽기
with open('medical_records.txt', 'r', encoding='utf-8') as file:
    first_line = file.readline().strip()    # 첫 번째 줄
    print(f"첫 번째 기록: {first_line}")
````

````python
# 3. 모든 줄을 리스트로 읽기
with open('medical_records.txt', 'r', encoding='utf-8') as file:
    all_lines = file.readlines()            # 모든 줄을 리스트로
    for i, line in enumerate(all_lines, 1): # 줄 번호와 함께
        print(f"{i}번째: {line.strip()}")
````

````python
# 4. for문으로 읽기 (가장 효율적)
with open('medical_records.txt', 'r', encoding='utf-8') as file:
    for line_num, line in enumerate(file, 1):  # 파일을 한 줄씩 순회
        record = line.strip()               # 현재 줄의 진료기록
        print(f"환자 {line_num}: {record}")
````

## ⚠️ 예외처리

**예외처리는** 프로그램 실행 중 발생할 수 있는 오류를 적절히 처리하여 시스템 안정성을 보장하는 방법입니다.



```python
# 예외가 발생하는 상황들
patient_file = open('nonexistent_file.txt', 'r')  # FileNotFoundError

result = 10 / 0                                 # ZeroDivisionError
```


### try-except 구문
**try-except 구문은 예외가 발생할 수 있는 코드를 안전하게 실행하기 위한 방법입니다.**

try 블록에서 오류가 발생하면 except 블록이 실행되어 프로그램이 중단되지 않습니다.

##### 기본 문법
```python
# 모든 예외 처리 (가장 기본)
try:
    # 오류가 발생할 수 있는 코드
    위험한_코드()
except:
    # 모든 오류를 잡음
    print("뭔가 오류가 발생했습니다")

```

```python
# 예외 정보 받기
try:
    위험한_코드()
except Exception as e:
    # 예외 객체를 e로 받아서 사용
    print(f"오류 발생: {e}")
```

```python
# 특정 예외 타입 지정
try:
    # 오류가 발생할 수 있는 코드
    위험한_코드()
except 예외타입:
    # 오류가 발생했을 때 실행할 코드
    오류_처리_코드()

```

````python
# 2. 예외 정보 받기
try:
    result = 10 / 0  # ZeroDivisionError 발생
except Exception as e:
    # 예외 객체를 e로 받아서 사용
    print(f"오류 발생: {e}")
````


#### 예외처리 전 vs 후 비교


**❌ 예외처리 전 - 위험한 코드**


````python
def unsafe_patient_input():
    """예외처리가 없는 위험한 환자 정보 입력"""
    patient_name = input("환자명: ")
    age_input = input("나이: ")
    patient_age = int(age_input)  # 숫자가 아니면 프로그램 중단!

    print(f"환자 등록: {patient_name}({patient_age}세)")
    return {"name": patient_name, "age": patient_age}

# 사용자가 "스물다섯"을 입력하면 → ValueError로 프로그램 종료

unsafe_patient_input()
````

**✅ 예외처리 후 - 안전한 코드**


````python
def safe_patient_input():
    """안전한 환자 정보 입력을 처리합니다"""
    try:
        patient_name = input("환자명: ")              # 환자 이름
        age_input = input("나이: ")                  # 나이 입력값
        patient_age = int(age_input)                 # 나이를 정수로 변환

        print(f"환자 등록: {patient_name}({patient_age}세)")
        return {"name": patient_name, "age": patient_age}

    except ValueError:
        print("나이는 숫자로 입력해주세요.")
        return None

# 사용자가 "스물다섯"을 입력해도 → 친절한 안내 메시지 출력

safe_patient_input()
````

### finally 블록
**finally 블록은 예외 발생 여부와 관계없이 항상 실행되는 코드 블록입니다.**

파일 닫기, 리소스 정리 등 반드시 실행되어야 하는 작업에 사용됩니다.

````python
def calculate_patient_bmi(weight, height):
    """환자 BMI를 계산하고 시스템 로그를 남깁니다"""

    try:
        print("BMI 계산을 시작합니다...")
        bmi = weight / ((height / 100) ** 2)  # BMI 계산
        print(f"BMI 계산 결과: {bmi:.2f}")
        return bmi

    except ZeroDivisionError:
        print("키는 0보다 커야 합니다.")
        return None

    except TypeError:
        print("숫자만 입력 가능합니다.")
        return None

    finally:
        print("=" * 30)
        print("BMI 계산 작업을 종료합니다.")  # 항상 실행
        print("시스템 로그에 기록되었습니다.")
````

````python
# ✅ 정상 케이스 - 올바른 체중과 키
result = calculate_patient_bmi(70, 175)
````

````python
# ❌ 오류 케이스 - 키가 0인 경우
result = calculate_patient_bmi(70, 0)
````

### FileNotFoundError


````python
def read_patient_file(patient_id):
    """환자 파일을 안전하게 읽습니다"""
    filename = f"patient_{patient_id}.txt"  # 환자별 파일명

    try:
        with open(filename, 'r', encoding='utf-8') as file:
            content = file.read()           # 파일 내용
            print("파일을 불러왔습니다.")
            return content

    except FileNotFoundError:
        print(f"환자 {patient_id}의 기록을 찾을 수 없습니다.")

        # 새 파일 생성 제안
        create_new = input("새 기록을 생성하시겠습니까? (y/n): ")
        if create_new.lower() == 'y':
            with open(filename, 'w', encoding='utf-8') as file:
                file.write(f"환자 {patient_id} 신규 기록\n")
            print("새 기록 파일이 생성되었습니다.")

        return None
````

### PermissionError


````python
def secure_medical_backup(source_file, backup_location):
    """의료 데이터를 안전하게 백업합니다"""
    try:
        with open(source_file, 'r', encoding='utf-8') as source:
            medical_data = source.read()    # 의료 데이터

        with open(backup_location, 'w', encoding='utf-8') as backup:
            backup.write(medical_data)

        print(f"백업 완료: {backup_location}")
        return True

    except PermissionError:
        print("백업 권한이 없습니다. 관리자에게 문의하세요.")
        return False
````

### ValueError


````python
def validate_patient_vitals(temperature, blood_pressure):
    """환자 생체징후를 검증합니다"""
    try:
        # 체온 검증
        temp_value = float(temperature)         # 체온값
        if temp_value < 35.0 or temp_value > 42.0:
            raise ValueError(f"비정상적인 체온: {temp_value}°C")

        # 혈압 검증
        if '/' not in blood_pressure:
            raise ValueError("혈압은 '수축기/이완기' 형식으로 입력하세요.")

        systolic, diastolic = blood_pressure.split('/')
        systolic_val = int(systolic)            # 수축기 혈압
        diastolic_val = int(diastolic)          # 이완기 혈압

        if systolic_val < 70 or systolic_val > 200:
            raise ValueError(f"비정상적인 수축기 혈압: {systolic_val}")

        print("생체징후가 정상적으로 검증되었습니다.")
        return {"temperature": temp_value, "systolic": systolic_val, "diastolic": diastolic_val}

    except ValueError as e:
        print(f"데이터 오류: {e}")
        return None
````

````python
# ✅ 정상 케이스 - 올바른 생체징후 데이터
result = validate_patient_vitals("36.5", "120/80")
````

````python
# ❌ 오류 케이스 1 - 잘못된 체온
result = validate_patient_vitals("44", "120/80")
````

````python
# ❌ 오류 케이스 2 - 잘못된 혈압 형식
result = validate_patient_vitals("36.5", "120-80")
````

## 6. 🛠️ 파일 관련 예외 처리
**파일 작업 시 자주 발생하는 예외 타입들을 이해하고 적절히 처리하는 것이 중요합니다.**


### 주요 파일 예외 타입

| 예외 타입 | 발생 원인 | 해결 방법 |
|----------|----------|----------|
| **FileNotFoundError** | 파일이 존재하지 않음 | 파일 경로 확인, 새 파일 생성 |
| **PermissionError** | 파일 접근 권한 없음 | 권한 요청, 다른 위치 사용 |
| **IsADirectoryError** | 파일 대신 폴더를 열려고 함 | 올바른 파일 경로 지정 |
| **UnicodeDecodeError** | 잘못된 인코딩으로 읽기 | UTF-8 인코딩 지정 |
| **OSError** | 운영체제 관련 오류 | 시스템 상태 확인 |
| **IOError** | 입출력 일반 오류 | 파일 상태 및 연결 확인 |

### 종합 파일 예외 처리


````python
def safe_file_read(filename):
    """안전한 파일 읽기 함수"""
    try:
        with open(filename, 'r', encoding='utf-8') as file:
            return file.read()

    except FileNotFoundError:
        print(f"'{filename}' 파일을 찾을 수 없습니다.")
        return None

    except PermissionError:
        print(f"'{filename}' 파일 접근 권한이 없습니다.")
        return None

    except UnicodeDecodeError:
        print(f"'{filename}' 파일 인코딩 오류입니다.")
        return None

    except Exception as e:
        print(f"예상치 못한 오류: {e}")
        return None

def safe_file_write(filename, content):
    """안전한 파일 쓰기 함수"""
    try:
        with open(filename, 'w', encoding='utf-8') as file:
            file.write(content)
        print(f"'{filename}' 저장 완료")
        return True

    except PermissionError:
        print(f"'{filename}' 쓰기 권한이 없습니다.")
        return False

    except Exception as e:
        print(f"저장 중 오류: {e}")
        return False
````

````python
# ✅ 정상 케이스 - 존재하는 파일 읽기 및 복사
content = safe_file_read('patient_records.txt')
if content:
    success = safe_file_write('patient_records_copy.txt', content)
    if success:
        print("파일 복사가 완료되었습니다.")
````

````python
# ❌ 오류 케이스 - 존재하지 않는 파일 읽기
content = safe_file_read('nonexistent_file.txt')
````

### 사용자 친화적 오류 메시지


````python
class PatientFileManager:
    def get_user_friendly_message(self, error_type, filename):
        """사용자 친화적인 오류 메시지를 생성합니다"""
        messages = {
            "FileNotFoundError": {
                "title": "📂 파일을 찾을 수 없습니다",
                "description": f"'{filename}' 파일이 없거나 삭제되었습니다.",
                "actions": ["파일 이름을 확인하세요", "새 파일을 생성하세요"]
            },
            "PermissionError": {
                "title": "🔒 파일 접근 권한이 없습니다",
                "description": f"'{filename}' 파일을 열 권한이 없습니다.",
                "actions": ["관리자에게 권한을 요청하세요", "다른 위치에 저장하세요"]
            }
        }
        # 에러 딕셔너리 반환
        return messages.get(error_type, {"title": "⚠️ 알 수 없는 오류", "description": "다시 시도하세요"})

    def safe_read_patient_file(self, filename):
        """환자 파일을 안전하게 읽습니다"""
        try:
            with open(filename, 'r', encoding='utf-8') as file:
                content = file.read()
                print(f"✅ '{filename}' 읽기 성공")
                return content

        except FileNotFoundError:
            msg = self.get_user_friendly_message("FileNotFoundError", filename)
            print(f"\n{msg['title']}")
            print(f"{msg['description']}")
            for action in msg['actions']:
                print(f"• {action}")
            return None

        except PermissionError:
            msg = self.get_user_friendly_message("PermissionError", filename)
            print(f"\n{msg['title']}")
            print(f"{msg['description']}")
            for action in msg['actions']:
                print(f"• {action}")
            return None

        except Exception as e:
            msg = self.get_user_friendly_message("Unknown", filename)
            print(f"\n{msg['title']}")
            print(f"{msg['description']}")
            print(f"• 기술적 오류: {e}")
            return None

````

````python
# ✅ 정상 케이스 - 존재하는 파일 읽기
manager = PatientFileManager()
content = manager.safe_read_patient_file("patient_records.txt")
if content:
    print("파일 읽기 성공!")
````

````python
# ❌ 오류 케이스 - 존재하지 않는 파일 읽기
content = manager.safe_read_patient_file("nonexistent_file.txt")
````
