### 매개변수(Parameter) - 함수 정의 시 사용

**매개변수는** 함수를 정의할 때 입력값을 받기 위해 설정하는 변수입니다.


````python
# name, age, department가 매개변수(Parameter)
def register_patient(name, age, department):
    """환자를 등록하는 함수 정의"""
    print(f"환자 등록: {name}님 ({age}세)")
    print(f"담당 과: {department}")
    print("등록이 완료되었습니다.")
````

````python
# height_cm, weight_kg가 매개변수(Parameter)
def calculate_bmi(height_cm, weight_kg):
    """BMI를 계산하는 함수 정의"""
    bmi = weight_kg / (height_cm / 100) ** 2
    return bmi
````

## 📤 반환값(return)

**return**은 함수에서 계산된 결과값을 호출한 곳으로 돌려주거나, 함수의 실행을 중단시키는 역할을 합니다.  
return은 반환받은 값을 다른 코드에 사용해야 할때 사용합니다.

image.png

### **return의 두 가지 용법**

#### 1. 값 반환 - 함수 결과를 돌려주기

````python
def calculate_bmi(height_cm, weight_kg):
    """BMI를 계산하여 반환합니다"""
    bmi = weight_kg / (height_cm / 100) ** 2
    return bmi

def classify_bmi_status(bmi):
    """BMI 값을 받아서 상태를 분류합니다"""
    if bmi < 18.5:
        return "저체중"
    elif bmi < 25:
        return "정상체중"
    elif bmi < 30:
        return "과체중"
    else:
        return "비만"
````

````python
# return 값을 다른 함수에 활용
patient_bmi = calculate_bmi(170, 70)  # BMI 계산
bmi_status = classify_bmi_status(patient_bmi)  # 계산된 BMI로 상태 분류
````

##### 여러개의 값 반환
함수에서 여러개의 값을 반환할때는 하나의 튜플로 묶여 반환됩니다.

````python
# 환자 데이터베이스
patient_database = [
    {"id": "P001", "name": "김환자", "temperature": 36.5, "pulse": 72, "systolic_bp": 120},
    {"id": "P002", "name": "이환자", "temperature": 38.2, "pulse": 85, "systolic_bp": 140},
    {"id": "P003", "name": "박환자", "temperature": 37.8, "pulse": 95, "systolic_bp": 160}
]
````

````python
def get_vital_signs(patient_id):
    """환자 ID로 데이터베이스에서 바이탈 사인을 조회합니다"""
    for patient in patient_database:
        if patient["id"] == patient_id:
            return patient["temperature"], patient["pulse"], patient["systolic_bp"] # 튜플로 반환
    return None, None, None  # 환자를 찾지 못한 경우
````

````python
# 바이탈 사인 조회
temp, pulse, bp = get_vital_signs("P002")
print(f"체온: {temp}°C, 맥박: {pulse}bpm, 혈압: {bp}mmHg")
````

#### 2. 함수 종료 - 실행 중단하기

````python
def check_adult_patient(age):
    """환자가 성인인지 확인합니다"""
    if age < 0:
        print("잘못된 나이 정보입니다.")
        return  # 여기서 함수 종료 (값 반환 없음)

    if age < 18:
        print("미성년자입니다. 보호자 동반이 필요합니다.")
        return  # 여기서 함수 종료 (값 반환 없음)

    print("성인 환자입니다.")
    print("진료를 진행하겠습니다.")
````

## 람다함수 (Lambda Functions)

**람다함수는** 간단한 함수를 한 줄로 정의할 수 있는 익명 함수입니다.

의료 데이터 처리에서 환자 목록 정렬, 조건부 필터링, 데이터 변환 등에 매우 유용합니다.

## 왜 람다함수가 필요할까?

- 간단한 작업을 위해 함수를 계속 만들기 번거로움
- 데이터 정렬/필터링에서의 편리함



### 🔥 람다함수 사용법
람다함수는 `lambda` 키워드를 사용해 간단하게 함수를 정의하며  `return` 키워드 없이 **표현식의 결과를 자동으로 리턴**합니다.


#### 기본 문법
```python
# 기본 형태
lambda 파라미터: 리턴할 표현식

# 예시
lambda x: x * 2
```


````python
# 일반 함수
def calculate_bmi(weight, height):
    return weight / (height ** 2)

# 람다함수로 변환
calculate_bmi_lamda = lambda weight, height: weight / (height ** 2)
````

````python
# 사용법은 동일
patient_bmi = calculate_bmi_lamda(70, 1.75)
print(f"환자 BMI: {patient_bmi:.1f}")
````

#### 표현식이란?
표현식(Expression)은 **값을 계산해서 결과를 만들어내는 코드**입니다.


````python
# 이것들이 모두 "표현식"입니다
x + 5           # 계산 표현식
"hello"         # 문자열 표현식
x > 10          # 비교 표현식
a if b else c   # 조건 표현식
````

### 실전 활용 예시



````python
# 환자 데이터베이스
patient_database = [
    {"id": "P001", "name": "김환자", "age": 45, "height": 170, "weight": 70},
    {"id": "P002", "name": "이환자", "age": 38, "height": 165, "weight": 85},
    {"id": "P003", "name": "박환자", "age": 52, "height": 175, "weight": 95}
]
````

````python
# 환자 진료 시스템 - 잘 설계된 함수들
def get_patient_info(patient_id):
    """환자 ID로 데이터베이스에서 기본 정보를 조회합니다"""
    for patient in patient_database:
        if patient["id"] == patient_id:
            return patient
    return None  # 환자를 찾지 못한 경우

def calculate_bmi(height_cm, weight_kg):
    """BMI를 계산합니다"""
    return weight_kg / (height_cm / 100) ** 2

def classify_weight_status(bmi):
    """BMI를 기준으로 체중 상태를 분류합니다"""
    if bmi < 18.5:
        return "저체중"
    elif bmi < 25:
        return "정상체중"
    elif bmi < 30:
        return "과체중"
    else:
        return "비만"

def create_health_report(patient_id):
    """환자의 건강 보고서를 생성합니다"""
    # 각 함수의 단일 책임을 활용
    patient_info = get_patient_info(patient_id)  # 데이터베이스에서 정보 조회
    if patient_info is None:
        return {"error": "환자를 찾을 수 없습니다"}

    bmi = calculate_bmi(patient_info["height"], patient_info["weight"])  # BMI 계산
    weight_status = classify_weight_status(bmi)  # 상태 분류

    return {
        "patient_name": patient_info["name"],
        "bmi": round(bmi, 1),
        "weight_status": weight_status
    }

````

````python
# 함수 사용
health_report = create_health_report("P002")
if "error" not in health_report:
    print(f"환자: {health_report['patient_name']}")
    print(f"BMI: {health_report['bmi']} ({health_report['weight_status']})")
else:
    print(health_report["error"])
````

### 실전 활용 예시


````python
def process_patient_vitals(
    patient_id: str,
    temperature: float,
    blood_pressure: list[int],
    symptoms: list[str]
) -> dict[str, any]:
    """환자 바이탈 사인을 처리하고 진단 정보를 반환합니다"""

    # 발열 여부 확인
    has_fever: bool = temperature >= 37.5

    # 고혈압 여부 확인 (수축기/이완기)
    is_hypertension: bool = blood_pressure[0] >= 140 or blood_pressure[1] >= 90

    # 응급도 판정
    emergency_level: str = determine_emergency_level(has_fever, is_hypertension, symptoms)

    return {
        "patient_id": patient_id,
        "has_fever": has_fever,
        "is_hypertension": is_hypertension,
        "emergency_level": emergency_level,
        "recommended_action": get_recommended_action(emergency_level)
    }

def determine_emergency_level(fever: bool, hypertension: bool, symptoms: list[str]) -> str:
    """응급도를 판정합니다"""
    critical_symptoms: list[str] = ["가슴통증", "호흡곤란", "의식저하"]

    if any(symptom in critical_symptoms for symptom in symptoms):
        return "위급"
    elif fever and hypertension:
        return "주의"
    else:
        return "일반"

def get_recommended_action(emergency_level: str) -> str:
    """응급도에 따른 권장 조치를 반환합니다"""
    actions: dict[str, str] = {
        "위급": "즉시 응급실 내원",
        "주의": "당일 진료 예약",
        "일반": "정기 검진 권장"
    }
    return actions.get(emergency_level, "추가 관찰 필요")
````

## 💊 실전 예제: 환자 건강 관리 시스템


````python
def get_patient_input() -> tuple[str, str, str, str]:
    """사용자로부터 환자 정보를 입력받습니다"""
    name = input("환자명을 입력하세요: ")
    phone = input("전화번호를 입력하세요 (예: 010-1234-5678): ")
    date = input("예약 날짜를 입력하세요 (예: 2024-03-15): ")
    time = input("예약 시간을 입력하세요 (예: 10:00): ")
    return name, phone, date, time

def validate_patient_name(name: str) -> bool:
    """환자명의 유효성을 검사합니다"""
    return len(name) >= 2

def validate_phone_number(phone: str) -> bool:
    """전화번호의 유효성을 검사합니다"""
    return len(phone) >= 10

def check_time_availability(time: str) -> bool:
    """예약 시간의 가능 여부를 확인합니다"""
    unavailable_times = ["09:00", "14:00", "16:00"]
    return time not in unavailable_times

def create_appointment_record(name: str, phone: str, date: str, time: str) -> dict[str, str]:
    """예약 기록을 생성합니다"""
    return {
        "name": name,
        "phone": phone,
        "date": date,
        "time": time,
        "status": "confirmed"
    }
````

````python
def process_appointment() -> dict[str, any]:
    """전체 예약 처리 과정을 관리합니다"""
    # 입력 받기
    name, phone, date, time = get_patient_input()

    # 유효성 검사
    if not validate_patient_name(name):
        return {"success": False, "message": "이름은 2글자 이상이어야 합니다."}

    if not validate_phone_number(phone):
        return {"success": False, "message": "전화번호가 올바르지 않습니다."}

    if not check_time_availability(time):
        return {"success": False, "message": f"{time}는 예약이 불가능한 시간입니다."}

    # 예약 생성
    appointment = create_appointment_record(name, phone, date, time)

    return {
        "success": True,
        "appointment": appointment,
        "message": f"{name}님의 {date} {time} 예약이 완료되었습니다."
    }
````

````python
# 함수 실행
result = process_appointment()

if result["success"]:
    print("✅", result["message"])
    print("예약 정보:", result["appointment"])
else:
    print("❌", result["message"])
````
