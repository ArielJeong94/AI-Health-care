# 3. 조건문

## 🤔 if문 - 상황에 따라 다르게

조건문은 **주어진 조건에 따라 다른 코드를 실행**하는 구문입니다.  
의료 현장에서는 환자의 상태나 검사 결과에 따라 다른 처치를 결정해야 하므로 조건문이 매우 중요합니다.

조건문은 조건식과 실행구문으로 이루어져 있습니다.

```python
# 조건문 기분 구문
if 조건식:
  # 조건식이 True인 경우 실행할 구문
  print("조건식이 True입니다")
```

## 🔵 bool 자료형과 조건문

bool(불린) 자료형은 참(True) 또는 거짓(False) 두 가지 값만 가집니다.  
조건문의 핵심이 되는 자료형입니다. **`조건식`에는 무조건 `bool`자료형이 들어가야 하기 때문입니다.**


````python
# ❌ 잘못된 예시 - 대입 연산자 사용
patient_age = 25

if patient_age = 18:  # SyntaxError!
    print("성인이 되었습니다!")
````

````python
# ✅ 올바른 예시 - 비교 연산자 사용
patient_age = 25

if patient_age == 18:
    print("성인이 되었습니다!")
````

### 조건문에서 bool 값 활용
bool 변수는 조건문에서 직접 사용할 수 있습니다. `== True`를 쓸 필요가 없습니다.


````python
# ❌ 잘못된 예시 - bool 변수에 불필요한 비교
is_emergency = True
if is_emergency == True:
  print("응급상황입니다!")
````

````python
# ✅ 올바른 예시 - bool 변수 직접 사용
if is_emergency:
    print("응급상황입니다!")
````

````python
# 환자 상태에 따른 처리
is_emergency = True
has_insurance = False

if is_emergency:  # True이면 실행
    print("응급실로 이동하세요!")

if not has_insurance:  # False이면 실행 (not 사용)
    print("보험 정보를 확인해주세요")
````

### 비교 연산의 결과는 bool
모든 비교 연산의 결과는 bool 자료형입니다.


````python
# 검사 결과 평가
blood_sugar = 180
is_diabetes = blood_sugar >= 140  # 결과: True

cholesterol = 220
is_high_cholesterol = cholesterol >= 240  # 결과: False

print(f"당뇨 진단: {is_diabetes}")
print(f"고콜레스테롤혈증: {is_high_cholesterol}")

````

## 🔍 조건식 연산자


### 비교 연산자
조건문에서 값들을 비교할 때 사용하는 연산자들입니다.


````python
# 환자 혈압 데이터
systolic = 140 # 환자의 수축기 혈압
normal_systolic = 120 # 정상 수축기 혈압 기준값

print(f"수축기 혈압 {systolic} > 정상 {normal_systolic}: {systolic > normal_systolic}")
print(f"혈압 정상여부: {systolic <= normal_systolic}")
print(f"고혈압 진단: {systolic >= 140}")
print(f"정상 혈압: {systolic == normal_systolic}")
print(f"혈압 이상: {systolic != normal_systolic}")
````

### 포함 연산자 (in, not in)
리스트나 문자열에 특정 값이 포함되어 있는지 확인할 때 사용합니다.

- `in`: 포함되어 있으면 True
- `not in`: 포함되지 않으면 True


````python
# 환자 알레르기 확인
patient_allergies = ["페니실린", "아스피린", "설파제"] # 환자의 알레르기 약물 목록
prescribed_drug = "페니실린" # 처방하려는 약물명

if prescribed_drug in patient_allergies:
    print(f"⚠️ 주의: {prescribed_drug}에 알레르기가 있습니다!")
else:
    print(f"✅ {prescribed_drug} 처방 가능합니다.")
````

````python
# 증상 확인
patient_symptoms = ["발열", "기침", "두통"] # 환자가 호소하는 증상 목록

if "호흡곤란" not in patient_symptoms:
    print("호흡곤란 증상은 없습니다.")
````

### 논리 연산자

논리 연산자는 **여러 조건을 조합**해서 더 복잡한 조건을 만들 때 사용합니다.

| 연산자 | 의미 | 설명 |
| --- | --- | --- |
| `and` | 그리고 | 모든 조건이 True일 때만 True |
| `or` | 또는 | 하나라도 True이면 True |
| `not` | 아니다 | True를 False로, False를 True로 |


#### and 연산자 - 모든 조건이 참일 때


````python
# 수술 가능 여부 판단 (모든 조건을 만족해야 함)
patient_age = 45 # 환자 나이
systolic = 130  # 환자의 수축기 혈압

if patient_age >= 18 and systolic < 180:
    print("✅ 수술 가능합니다.")
else:
    print("❌ 수술 전 추가 검사가 필요합니다.")
````

#### or 연산자 - 하나라도 참이면


````python
# 응급실 방문 권고 (하나라도 해당되면 응급상황)
chest_pain = True  # 가슴 통증 여부
difficulty_breathing = False  # 호흡곤란 여부
severe_headache = False  # 심한 두통 여부

if chest_pain or difficulty_breathing or severe_headache:
    print("🚨 즉시 응급실로 오세요!")
else:
    print("일반 진료 예약을 잡으시기 바랍니다.")
````

#### not 연산자 - 반대 조건


````python
# 처방 금기 확인
is_pregnant = False  # 임신 여부
is_allergic = False  # 해당 약물 알레르기 여부

if not is_pregnant and not is_allergic:
    print("✅ 해당 약물 처방 가능합니다.")
````

````python
# 백신 접종 가능 여부
has_fever = True # 발열 여부

if not has_fever:
    print("백신 접종이 가능합니다.")
else:
    print("발열이 있어 백신 접종을 연기합니다.")

````

## 3️⃣ 삼항 연산자 (Ternary Operator)

삼항 연산자는 **간단한 if-else 구문을 한 줄로 표현**하는 방법입니다.  
조건문 코드를 간결하게 작성할 수 있습니다.


### 🔀 삼항 연산자 기본 구조

```python
변수 = 값1 if 조건 else 값2
```

조건이 **True**이면 `값1`, **False**이면 `값2`를 변수에 저장합니다.


#### 기본 if-else vs 삼항 연산자


````python
# 기본 if-else 구문
patient_age = 17

if patient_age >= 18:
    age_group = "성인"
else:
    age_group = "미성년자"

print(f"환자 분류: {age_group}")

````

````python
# 삼항 연산자 사용
patient_age = 17
age_group = "성인" if patient_age >= 18 else "미성년자"

print(f"환자 분류: {age_group}")

````

# 추가 검사 필요 여부
additional_test = "필요" if systolic >= 140 else "불필요"

### 삼항 연산자 사용 팁

    result = "위험"

## 🔀 조건문 처리 방식 비교

| 방식 | 사용 시점 | 특징 | 예시 |
| --- | --- | --- | --- |
| **다중 조건** (if-elif-else) | 하나만 선택해야 할 때 | 첫 번째 참인 조건만 실행 | 혈압 등급 분류, 체중 등급 판정 |
| **독립 조건** (if-if-if) | 각각 별도로 확인할 때 | 모든 조건을 개별 확인 | 여러 증상 체크, 알레르기 확인 |
| **중첩 조건** (if 안의 if) | 단계별 확인이 필요할 때 | 조건을 만족해야 다음 단계 진행 | 응급도 평가 → 진료과 배정 |

````python
# 통합 환자 데이터
patient = {
    "name": "김환자",
    "age": 45,
    "symptoms": ["발열", "기침", "두통"],
    "insurance": True,
    "emergency_level": "응급",
    "department": None,
    "doctor": None
}

patient
````

### 독립 조건 - if - if -if

````python
# 1. 독립 조건: 각 증상별 개별 처방 (모두 실행 가능)
print("=== 증상별 개별 처방 ===")
if "발열" in patient["symptoms"]:
    print("해열제를 처방합니다.")

if "기침" in patient["symptoms"]:
    print("기침약을 처방합니다.")

if "두통" in patient["symptoms"]:
    print("진통제를 처방합니다.")
````

### 다중 조건 if-elif-else

````python
# 2. 다중 조건: 나이에 따른 진료과 배정 (하나만 선택)
print("=== 나이별 진료과 배정 ===")
if patient["age"] < 18:
    patient["department"] = "소아과"
elif patient["age"] < 65:
    patient["department"] = "내과"
else:
    patient["department"] = "노인의학과"

print(f"배정 진료과: {patient['department']}")
````

### 중첩 조건 - if 안에 if

````python
# 3. 중첩 조건: 단계별 진료 절차
print("=== 응급 환자 처리 절차 ===")

# 1단계: 응급도 확인
if patient["emergency_level"] == "응급":
    print("응급 환자입니다.")

    # 2단계: 보험 확인
    if patient["insurance"]:
        print("보험 적용으로 즉시 치료 시작합니다.")

        # 3단계: 진료과에 따른 담당의 배정
        if patient["department"] == "내과":
            patient["doctor"] = "김내과"
            print(f"담당의: {patient['doctor']} 배정완료")
        elif patient["department"] == "소아과":
            patient["doctor"] = "이소아"
            print(f"담당의: {patient['doctor']} 배정완료")
        else:
            print("담당의 배정 대기 중입니다.")

    else:
        print("보험 미적용이지만 응급이므로 치료 후 수납하겠습니다.")

else:
    print("일반 환자입니다. 접수 후 대기해주세요.")
````

## 💊 실전 예제: 환자 입원 승인 시스템



````python
patient = {
    "name": "이환자",  # 환자 이름
    "temperature": 37.2, # 체온 (섭씨)
    "bed_preference": "일반실",  # 희망 병실 유형
    "insurance_type": "건강보험",  # 보험 종류
    "emergency_contact": "010-1234-5678"  # 응급 연락처
}

# 병원 상황 (남은 병상 수)
hospital_status = {
    "일반실": 2,
    "특실": 5,
    "중환자실": 0
}
````

````python
print("=== 입원 가능 여부 판정 ===")

# 1단계: 바이탈 사인 확인
if patient["temperature"] < 38.0:
    print("체온 정상 범위입니다.")

    # 2단계: 병상 확인
    if hospital_status[patient["bed_preference"]] > 0:
        print(f"✅ {patient['bed_preference']} 입원 가능합니다.")

        # 3단계: 보험 확인
        if patient["insurance_type"] == "건강보험":
            print("건강보험 적용으로 입원 수속을 진행합니다.")
        else:
            print("실비보험 확인 후 입원 수속을 진행합니다.")

    else:
        print(f"❌ {patient['bed_preference']}에 여유 병상이 없습니다.")
        print("다른 병실을 확인하거나 대기 등록하시겠습니까?")

else:
    print("⚠️ 발열로 인해 격리병실 배정이 필요합니다.")
````
