## 스칼라 (Scalar) - 하나의 숫자

환자의 **개별 정보**를 나타냅니다.

````python
# 환자 한명의 정보들 (각각이 스칼라)
patient_age = 45        # 나이 (세)
patient_weight = 70.5   # 체중 (kg)
patient_height = 175    # 키 (cm)
patient_bp_high = 120   # 수축기 혈압 (mmHg)
patient_bp_low = 80     # 이완기 혈압 (mmHg)

print("=== 환자 개별 정보 (스칼라) ===")
print(f"나이: {patient_age}세")
print(f"체중: {patient_weight}kg")
print(f"키: {patient_height}cm")
print(f"혈압: {patient_bp_high}/{patient_bp_low}mmHg")
````

# NumPy란?

NumPy는 AI와 머신러닝의 **핵심 도구**입니다. 의료 현장에서 환자 데이터 분석, 의료 영상 처리, 진단 예측 모델 등을 구현할 때 반드시 사용되는 라이브러리입니다.

파이썬의 기본 리스트로는 수백만 개의 데이터를 빠르게 처리할 수 없지만, NumPy는 **벡터화 연산**을 통해 수천 배 빠른 연산이 가능합니다.

🏥 **의료 현장 예시**: 환자 1만 명의 혈압 데이터 분석
- 파이썬 리스트: 10분 소요
- NumPy 배열: 1초 소요

속도가 생명인 의료 AI에서 NumPy는 필수입니다!

### Numpy 연습 사이트: https://www.machinelearningplus.com/python/101-numpy-exercises-python/

````python
import numpy as np
print(f"NumPy 버전: {np.__version__}")
````

# 2. 파이썬 리스트 vs NumPy 배열


같은 작업을 파이썬 리스트와 NumPy로 각각 수행해서 **속도 차이**를 직접 체험해보겠습니다.

**시나리오**: 환자 10,000명의 키를 센티미터에서 미터로 변환

````python
import time
import random

# 10,000명의 환자 키 데이터 생성 (150-200cm 사이 랜덤)
num_patients = 10000
heights_list = [random.uniform(150, 200) for _ in range(num_patients)]   # 파이썬 리스트
heights_numpy = np.array(heights_list)                                   # Numpy 배열

print(f"환자 수: {num_patients:,}명")
print(f"키 데이터 예시: {heights_list[:5]} (cm)")
````

## 방법 1: 파이썬 리스트 + for 반복문

````python
# 파이썬 리스트로 cm → m 변환
start_time = time.time()

heights_meter_list = []
for height in heights_list: # 10000번 반복
    heights_meter_list.append(height / 100)  # cm를 m로 변환

python_time = time.time() - start_time

print(f"파이썬 리스트 방법: {python_time:.6f}초")
print(f"결과 예시: {heights_meter_list[:5]} (m)")
````

## 방법 2: NumPy 벡터화 연산

````python
# NumPy로 cm → m 변환 (벡터화 연산)
start_time = time.time()

heights_meter_numpy = heights_numpy / 100  # 전체 배열을 한 번에 연산!

numpy_time = time.time() - start_time

print(f"NumPy 벡터화 방법: {numpy_time:.6f}초")
print(f"결과 예시: {heights_meter_numpy[:5]} (m)")
````

## ⚡ 성능 비교 결과

````python
speedup = python_time / numpy_time

print("=== 성능 비교 결과 ===")
print(f"파이썬 리스트: {python_time:.6f}초")
print(f"NumPy 배열:    {numpy_time:.6f}초")
print(f"NumPy가 {speedup:.1f}배 더 빠름! 🚀")

# 결과 검증 (두 방법이 같은 결과를 내는지 확인)
# np.allclose(): 두 배열이 "거의 같은지" 확인하는 함수
print(f"\n결과 일치 여부: {np.allclose(heights_meter_list, heights_meter_numpy)}")
````

# 4. 배열의 속성과 데이터 추출

## shape, ndim, dtype으로 데이터 구조 파악

데이터를 다루기 전에 **데이터의 구조**를 파악하는 것이 중요합니다.

- `ndim`: 배열의 차원 수 - 벡터(1차원)인지 행렬(2차원)인지 구분  
- `dtype`: 데이터 타입 - 정수형, 실수형, 문자형 등 확인
- `size`: 전체 요소 개수 - 총 데이터 포인트 수 파악

````python
# 1차원 배열 (벡터): 환자 한 명의 정보
patient_vector = np.array([45, 70.5, 175, 120, 80])
print("=== 1차원 배열 (벡터) ===")
print(f"환자 정보: {patient_vector}")
print(f"배열 형태: {patient_vector.shape}")
print(f"배열 차원: {patient_vector.ndim}")
print(f"데이터 타입: {patient_vector.dtype}")
````

## 인덱싱과 슬라이싱으로 필요한 정보 추출

환자별, 특성별로 필요한 데이터만 선별해서 추출할 수 있습니다.

- **인덱싱**: 특정 위치의 하나의 값 추출 (`array[행, 열]`)
- **슬라이싱**: 연속된 여러 값을 범위로 추출 (`array[시작:끝]`)
- **전체 선택**: 콜론(`:`) 사용으로 해당 차원의 모든 데이터 선택

````python
patients_matrix = np.array([
    [45, 70.5, 175, 120, 80],   # 환자1
    [32, 55.0, 162, 110, 75],   # 환자2
    [67, 82.3, 168, 140, 90],   # 환자3
    [28, 48.7, 155, 105, 70],   # 환자4
])

print("=== 여러 환자의 특성 묶음 (행렬) ===")
print("형태: [나이, 체중, 키, 수축기BP, 이완기BP]")
print(f"\n{patients_matrix}")
````

### 인덱싱
데이터 한개 추출

````python
print("=== 특정 환자 정보 추출 ===")
first_patient = patients_matrix[0]  # 첫 번째 환자
last_patient = patients_matrix[-1]  # 마지막 환자

print(f"첫 번째 환자: {first_patient}")
print(f"마지막 환자: {last_patient}")

# 특정 환자의 특정 정보
kim_age = patients_matrix[0, 0]      # 첫 번째 환자의 나이
park_weight = patients_matrix[-1, 1]  # 마지막 환자의 체중

print(f"\n첫 번째 환자의 나이: {kim_age}세")
print(f"마지막 환자의 체중: {park_weight}kg")
````

### 슬라이싱
여러 데이터를 범위로 추출

````python
# 슬라이싱: 여러 데이터를 한 번에
print("=== 여러 데이터 추출 (슬라이싱) ===")

# 모든 환자의 나이 (첫 번째 열)
all_ages = patients_matrix[:, 0]
print(f"모든 환자의 나이: {all_ages}")

# 모든 환자의 체중 (두 번째 열)
all_weights = patients_matrix[:, 1]
print(f"모든 환자의 체중: {all_weights}")

# 처음 3명 환자의 모든 정보
first_three_patients = patients_matrix[:3, :]
print(f"\n처음 3명 환자:\n{first_three_patients}")

# 모든 환자의 혈압 정보 (마지막 2개 열)
all_blood_pressure = patients_matrix[:, -2:]
print(f"\n모든 환자의 혈압:\n{all_blood_pressure}")
````

## 조건부 인덱싱 (Boolean Indexing)

특정 조건을 만족하는 원소들을 한번에 선택합니다.

- **Boolean 마스크**: 조건 결과로 True/False 배열 생성
  
  `ages >= 50` → `[False, True, True, False, True]`
- **마스크 적용**: True인 위치의 데이터만 추출
  
  `patients[ages >= 50]` → 조건 만족하는 모든 환자 정보


````python
# 조건부 선별: 고혈압 환자 찾기 (수축기 혈압 >= 140)
systolic_bp = patients_matrix[:, 3]  # 수축기 혈압 (4번째 열)
high_bp_mask = systolic_bp >= 130    # 조건 마스크

print("=== 고혈압 환자 찾기 ===")
print(f"모든 환자의 수축기 혈압: {systolic_bp}")
print(f"고혈압 여부 (>=130): {high_bp_mask}")

# 고혈압 환자의 모든 정보
high_bp_patients = patients_matrix[high_bp_mask]
print(f"\n고혈압 환자들:\n{high_bp_patients}")

# 나이 40세 이상 환자들
ages = patients_matrix[:, 0]
elderly_patients = patients_matrix[ages >= 40]
print(f"\n40세 이상 환자들:\n{elderly_patients}")
````

# 6. 기초 통계 함수

## 의료 데이터 분석의 핵심: 환자군 통계 분석

**기초 통계 함수**는 의료진이 환자군의 특성을 파악하고 임상 결정을 내리는 데 필수적인 도구입니다.   
환자들의 생체 지표에서 최솟값, 최댓값, 평균값을 찾아 정상 범위를 설정하고 이상 케이스를 식별합니다.



````python
# 10명 환자의 수축기 혈압 데이터 (mmHg)
blood_pressure = np.array([120, 135, 118, 142, 128, 156, 110, 139, 125, 148])

print("=== 환자 혈압 기본 통계 ===")
print(f"환자 혈압 데이터: {blood_pressure}")
print(f"환자 수: {len(blood_pressure)}명")

````

## 기본 통계 함수


````python
# 기본 통계 함수들
min_bp = np.min(blood_pressure)         # 최솟값
max_bp = np.max(blood_pressure)         # 최댓값
mean_bp = np.mean(blood_pressure)       # 평균값
std_bp = np.std(blood_pressure)         # 표준편차

print(f"\n📊 혈압 분석 결과:")
print(f"최저 혈압: {min_bp} mmHg")
print(f"최고 혈압: {max_bp} mmHg")
print(f"평균 혈압: {mean_bp:.1f} mmHg")
print(f"표준편차: {std_bp:.1f} mmHg")

# 분석 결과 분류
if min_bp < 90:
    print(f"\n⚠️  저혈압 환자 발견: {min_bp} mmHg")
else:
    print(f"\n✅ 저혈압 환자 없음 (최저: {min_bp} mmHg)")

if max_bp >= 140:
    print(f"\n⚠️  고혈압 환자 발견: {max_bp} mmHg")
else:
    print(f"\n✅ 고혈압 환자 없음 (최고: {max_bp} mmHg)")

````

## argmin과 argmax


argmin과 argmax는 값이 아니라 **위치(인덱스)를** 찾아주는 함수입니다.  
의료 데이터에서 "어떤 환자가 가장 높은/낮은 수치를 가지고 있는지"를 빠르게 식별할 때 매우 유용합니다.

- **argmin**: 배열에서 최솟값이 위치한 인덱스 반환
- **argmax**: 배열에서 최댓값이 위치한 인덱스 반환  


````python
## argmin과 argmax: 극값 환자 찾기

# 극값의 위치(인덱스) 찾기
min_index = np.argmin(blood_pressure)    # 최솟값 위치
max_index = np.argmax(blood_pressure)    # 최댓값 위치

print("=== 극값 환자 식별 ===")
print(f"최저 혈압 환자: {min_index+1}번째 환자 ({blood_pressure[min_index]} mmHg)")
print(f"최고 혈압 환자: {max_index+1}번째 환자 ({blood_pressure[max_index]} mmHg)")

# 환자 ID와 연결
patient_ids = np.array(['P001', 'P002', 'P003', 'P004', 'P005',
                       'P006', 'P007', 'P008', 'P009', 'P010'])

print(f"\n👤 환자 식별:")
print(f"최저 혈압: {patient_ids[min_index]} 환자 ({blood_pressure[min_index]} mmHg)")
print(f"최고 혈압: {patient_ids[max_index]} 환자 ({blood_pressure[max_index]} mmHg)")

print(f"\n📋 특별 관리 대상:")
print(f"• {patient_ids[min_index]} 환자: 저혈압 모니터링 필요")
print(f"• {patient_ids[max_index]} 환자: 고혈압 치료 우선순위")
````

## 축(axis) 개념

axis는 다차원 배열에서 "어느 방향으로 연산을 수행할지"를 지정하는 파라미터입니다.

- **axis=0**: 세로 방향 → 열별 통계 (각 생체지표별 전체 환자 평균)
- **axis=1**: 가로 방향 → 행별 통계 (각 환자별 모든 지표 평균)
- **의미**: axis=0은 "환자군 현황 파악", axis=1은 "개별 환자 종합 평가"

````python
## 다차원 배열에서의 통계

# 여러 환자의 다양한 생체 지표
# 행: 환자, 열: [나이, 수축기혈압, 혈당]
patients_data = np.array([
    [45, 120, 95],    # 환자1
    [67, 156, 180],   # 환자2
    [32, 110, 88],    # 환자3
    [58, 142, 102],   # 환자4
    [29, 118, 92]     # 환자5
])

features = ['나이', '수축기혈압', '혈당']
````

````python
print("=== 다차원 의료 데이터 분석 ===")
print(f"환자 데이터 (5명 × 3지표):")
print(f"{'환자':<5} {'나이':<5} {'혈압':<8} {'혈당':<5}")
print("-" * 25)
for i, data in enumerate(patients_data, 1):
    print(f"P{i:03d}  {data[0]:<5} {data[1]:<8} {data[2]:<5}")

# axis=0: 열 방향 (환자들을 비교) → 각 지표별 통계
# axis=1: 행 방향 (지표들을 비교) → 각 환자별 통계

print(f"\n📊 지표별(열방향) 통계 (axis=0):")
min_by_feature = np.min(patients_data, axis=0)   # ndarray
max_by_feature = np.max(patients_data, axis=0)   # ndarray
mean_by_feature = np.mean(patients_data, axis=0) # ndarray

for i, feature in enumerate(features):
    print(f"{feature}: 최소 {min_by_feature[i]:.0f}, 최대 {max_by_feature[i]:.0f}, 평균 {mean_by_feature[i]:.1f}")

print(f"\n👥 환자별(행방향) 평균 점수 (axis=1):")
patient_means = np.mean(patients_data, axis=1)
for i, mean_score in enumerate(patient_means, 1):
    print(f"환자 P{i:03d}: {mean_score:.1f}")

````

# 7. 내적과 행렬곱

## AI 예측의 핵심: 환자 정보 × 가중치 벡터

**내적(Dot Product)**은 AI가 예측을 만드는 핵심 연산입니다. 환자의 정보와 AI가 학습한 가중치를 곱해서 예측값을 계산합니다.

## 두 벡터의 내적(dot product)

````python
# 간단한 내적 예시
vector_a = np.array([1, 2, 3])
vector_b = np.array([4, 5, 6])

print("=== 내적 기본 개념 ===")
print(f"벡터 A: {vector_a}")
print(f"벡터 B: {vector_b}")

# 내적 계산: 대응 원소끼리 곱해서 모두 더하기
dot_manual = vector_a[0] * vector_b[0] + vector_a[1] * vector_b[1] + vector_a[2] * vector_b[2]

print(f"\n수동 계산: {vector_a[0]}×{vector_b[0]} + {vector_a[1]}×{vector_b[1]} + {vector_a[2]}×{vector_b[2]} = {dot_manual}")
````

````python
# NumPy에서 내적 계산하는 3가지 방법
dot_numpy = np.dot(vector_a, vector_b)      # 1. np.dot() 함수 - 가장 기본적
dot_operator = vector_a @ vector_b          # 2. @ 연산자 (Python 3.5+) - 최신 트렌트
dot_method = vector_a.dot(vector_b)         # 3. .dot() 메소드 - 비추천

print(f"np.dot(): {dot_numpy}")
print(f"@ 연산자: {dot_operator}")
print(f".dot() 메소드: {dot_method}")
````

## 의료 예측 시뮬레이션: 당뇨병 위험도 예측

````python
# 당뇨병 위험도 예측 시나리오
# 환자 정보: [나이, 체중, BMI]
# AI가 학습한 가중치 (가상의 값)

patient_info = np.array([45, 85, 25])  # 환자 데이터
weights = np.array([0.02, 0.015, 0.08])  # 학습된 가중치

print("=== 당뇨병 위험도 예측 ===")
print(f"환자 데이터: {patient_info}")
print(f"학습된 가중치: {weights}")

# 내적으로 위험도 점수 계산
risk_score = np.dot(patient_info, weights)

print(f"\n=== 내적 계산 ===")
print(f"45 × 0.02 + 85 × 0.015 + 25 × 0.08")
print(f"= 0.9 + 1.275 + 2.0")
print(f"= {risk_score:.3f} (위험도 점수)")

# 위험도 해석
if risk_score < 2.0:
    risk_level = "낮음"
elif risk_score < 4.0:
    risk_level = "보통"
elif risk_score < 6.0:
    risk_level = "높음"
else:
    risk_level = "매우 높음"

print(f"\n📊 당뇨병 위험도: {risk_level}")
````

## 여러 환자 동시 예측: 행렬곱 활용

````python
# 여러 환자 데이터를 3개 특성으로 구성 (나이, 체중, BMI)
patients_data = np.array([
    [45, 85, 25],   # 환자1
    [32, 60, 22],   # 환자2
    [67, 90, 28],   # 환자3
])

print("=== 전체 환자 당뇨병 위험도 예측 ===")
print(f"환자 데이터 (3명 × 3특성):\n{patients_data}")
print(f"가중치 벡터: {weights}")

# 환자의 위험도 수동 계산
risk_scores_method = []
for i in range(len(patients_data)):
    # 각 환자의 가중합 계산
    risk_score = (patients_data[i, 0] * weights[0] +
                  patients_data[i, 1] * weights[1] +
                  patients_data[i, 2] * weights[2])
    risk_scores_method.append(risk_score)
    print(f"환자 {i+1}: {patients_data[i, 0]} × {weights[0]} + {patients_data[i, 1]} × {weights[1]} + {patients_data[i, 2]} × {weights[2]} = {risk_score}")


print(f"\n=== 수동 계산 예측 결과 ===")
print(f"결과: {np.round(risk_scores_method, 3)}")


````

````python
# 행렬곱으로 모든 환자의 위험도 한 번에 계산
all_risk_scores = np.dot(patients_data, weights)

print(f"\n=== 예측 결과 ===")
print(f"위험도 점수들: {np.round(all_risk_scores, 3)}")

print(f"\n=== 환자별 위험도 분석 ===")
for i, score in enumerate(all_risk_scores, 1):
    if score < 2.0:
        risk_level = "낮음"
        emoji = "💚"
    elif score < 4.0:
        risk_level = "보통"
        emoji = "💛"
    elif score < 6.0:
        risk_level = "높음"
        emoji = "🧡"
    else:
        risk_level = "매우 높음"
        emoji = "❤️"

    print(f"{emoji} 환자{i}: {score:.3f} ({risk_level})")
````

## 행렬 전치(Transpose)

행과 열의 위치를 서로 바꾸는 연산입니다.

- **위치 변환**: (행, 열) → (열, 행)으로 좌표가 바뀜
  - 원본 `[0, 1]` 위치 → 전치 후 `[1, 0]` 위치
  - 원본 `[2, 0]` 위치 → 전치 후 `[0, 2]` 위치
- **형태 변환**: (m×n) 행렬 → (n×m) 행렬로 변환
  - 예: (3×4) → (4×3), (5×2) → (2×5)
- **사용 목적**: 행렬곱에서 shape 맞추기, 데이터 구조 변경 (환자별 -> 특성별)
- **표기법**: `array.T` 또는 `np.transpose(array)`

````python
# 행렬 전치: Shape 맞추기를 위한 활용
print("=== 행렬곱을 위한 전치 활용 ===")

# 환자 데이터: 2명, 3개 특성 (나이, 체중, BMI)
patients_data = np.array([
    [45, 70, 25],  # 환자1
    [32, 55, 22]   # 환자2
])

# 가중치: 2개 값 (환자별 가중치)
weights = np.array([0.8, 1.2])

print(f"환자 데이터 shape: {patients_data.shape}")  # (2, 3)
print(f"가중치 shape: {weights.shape}")             # (2,)
````

````python
# ❌ (2×3) × (2,) → 불가능 (열 개수 ≠ 행 개수)

result = np.dot(patients_data, weights)
````
